### a)

Qual o posto da aplicação composta

$$
L_Y\circ R_{X,d}?
$$

Pela questão 2(b) temos: $L_Y = V_y$, onde a matriz $V_y$ é uma Vandermonde aplicada nos pontos de $Y$ e $R_{X,d} = V_x^{+}$, a pseudoinversa da matriz de Vandermonde associada a $L_X$.


Portanto, a composição das duas transformações lineares vai ser:


$$
L_Y\circ R_{X,d} = V_YV_x^{+}
$$


A partir disso, temos que o posto do produto de duas matrizes nunca excede o posto de cada uma delas individualmente, i.e.:

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\operatorname{rank}(V_Y),\operatorname{rank}(V_X^+)\}.
$$

Mas o posto da pseudo inversa de $V_x$ é igual ao posto da própria $V_x$. Já que $ m \geq d+1 $, o posto de $V_x$ será

$$
\operatorname{rank}(V_X)=\min\{m,d+1\}=d+1.
$$

Logo:

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\operatorname{rank}(V_Y),d+1\}.
$$


E como 

$$
\operatorname{rank}(V_Y)=\min\{n,d+1\},
$$

temos

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\min\{n,d+1\},d+1\}.
$$

$$
\Rightarrow
rank(V_y V_x^+) \leq min\{n, d + 1\}

$$

Vamos além! Como $m \geq d+1$, a transformação 

$$
R_{X,d}:\mathbb R^m\to P_d
$$


é sobrejetiva, i.e.:

$$
\operatorname{Im}(R_{X,d})=P_d.
$$

Logo,

$$
\operatorname{rank}(L_Y\circ R_{X,d})
=
\dim(\operatorname{Im}(L_Y\circ R_{X,d}))
=
\dim(\operatorname{Im}(L_Y))
=
\operatorname{rank}(L_Y).
$$

Como

$$
\operatorname{rank}(L_Y)=\min\{n,d+1\},
$$

concluímos que

$$
\boxed{
\operatorname{rank}(L_Y\circ R_{X,d})
=
\min\{n,d+1\}.
}

### b)

Explique porque a composta acima não depende da base escolhida para $P_d$.

Fixe uma base b de $P_d$. Queremos provar que para qualquer outra base $C$ de $P_d$, a composta independe da base escolhida, isto é:

$$
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
=
[L_Y]_{\mathcal{C}}[R_{x,d}]_{\mathcal{C}}
$$

Se $p$ é um polinômio em $P_d$, podemos denotar o mesmo polinômios em duas bases:

$$
[p]_{\mathcal{B}} , [p]_{\mathcal{C}}
$$

Logo, vai existir uma matriz $S$ de mudança de base tal que:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Vamos encontrar a matriz de $L_Y$ na base C. Sabemos que:

$$
L_y(p) = [L_Y]_{\mathcal{B}}[p]_{\mathcal{B}}
$$

Entretanto, podemos reescrever isso como:

$$
L_y(p) = [L_Y]_{\mathcal{B}}S[p]_{\mathcal{C}}
$$

Como a transformação $L_Y$ aplicada no mesmo polinômio p mas na base C é:

$$
L_y(p) = [L_Y]_{\mathcal{C}}[p]_{\mathcal{C}}
$$

Podemos concluir que:

$$
[L_Y]_{\mathcal{C}} = [L_Y]_{\mathcal{B}}S
$$

Vamos olhar agora para $R_{x,d}$ na base C. Sabemos que para cada $b$, o vetor $[R_{x,d}]b_{\mathcal{B}}$ denota os coeficientes do polinômio minimizador associado a $b$ na base $B$. Entretanto, o vetor $[R_{x,d}]b_{\mathcal{B}}$ irá retornar os coeficientes do mesmo polinômio minimizador, mas na base $C$. Como as coordenadas do mesmo polinômio em bases diferentes se relacionam por:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Então, temos que:

$$
[R_{x,d}b]_{\mathcal{B}} = S[R_{x,d}b]_{\mathcal{C}}
$$

Invertendo a matriz $S$, concluímos que:

$$
[R_{x,d}b]_{\mathcal{C}} = S^{-1}[R_{x,d}b]_{\mathcal{C}}
$$

Agora que temos as matrizes de $L_Y$ e $R_{x,d}$ na base $C$, vamos compôr essas transformações:

$$
L_Y\circ R_{X,d} 
=
[L_Y]_{\mathcal{C}} [R_{x,d}b]_{\mathcal{C}}
=
([L_Y]_{\mathcal{B}}S) (S^{-1}[R_{x,d}b]_{\mathcal{C}})
=
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
$$

Assim, conseguimos provar que a matriz da composta independe da base.
