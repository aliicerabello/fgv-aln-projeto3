## Questão 3. Interpolação e Extrapolação

Considere

$$
X=\operatorname{linspace}(-1,1,m),
$$

ou seja, $m$ pontos igualmente espaçados em $[-1,1]$, e, analogamente,

$$
Y=\operatorname{linspace}(-1,1,n).
$$

Continuamos supondo $m\geq d+1$.

### a)

Qual o posto da aplicação composta

$$
L_Y\circ R_{X,d}?
$$

Pela questão 2(b) temos: $L_Y = V_y c$ e $R_{X,d} = V_x^{+}$
Portanto:
$$
L_Y\circ R_{X,d} = V_y V_x^{+}
$$
E podemos deduzir o seguinte:
$$
rank(V_y V_x^+) \leq min\{rank(v_y), V_x^+\}
$$
Mas o posto da pseudo inversa de $v_x$ é igual ao da própria:
$$
rank(V_y V_x^+) \leq min\{rank(v_y), d + 1\}
$$
E como $rank(v_y) = min\{n, d + 1\}$:
$$
rank(V_y V_x^+) \leq min\{min\{n, d + 1\}, d + 1\} 
\rightarrow
rank(V_y V_x^+) \leq min\{n, d + 1\}
$$
