## Questão 2. Regressão polinomial

Suponha que $m\geq d+1$.

Defina a função

$$
R_{X,d}:\mathbb{R}^m\to P_d
$$

que associa a cada vetor $b\in\mathbb{R}^m$ o polinômio $p\in P_d$ que melhor se ajusta aos valores $b_i$ nos pontos $x_i\in X$ no sentido de mínimos quadrados, ou seja,

$$
R_{X,d}(b)
=
\operatorname{argmin}_{p\in P_d}
\|L_X(p)-b\|_2^2.
$$



### a)

Mostre que $R_{X,d}$ é uma inversa à esquerda de $L_X$, ou seja,

$$
R_{X,d}\circ L_X=\operatorname{id}_{P_d}.
$$



O QUE EU QUERO QUE O BERNARDO SAIBA:


Queremos mostrar que:

$$
R_{X,d}\circ L_X = \operatorname{id}_{P_d}.
$$

Ou seja, para todo polinômio $p\in P_d$, queremos provar que

$$
R_{X,d}(L_X(p))=p.
$$

Seja $p\in P_d$. Aplicando $L_X$ a $p$, vamos obter o vetor

$$
L_X(p)
=
(p(x_1),p(x_2),\dots,p(x_m)).
$$

Vamos chamar esse vetor de

$$
y=L_X(p)\in \mathbb{R}^m.
$$

Pela definição de $R_{X,d}$, temos

$$
R_{X,d}(y)
=
\operatorname{argmin}_{q\in P_d}
\|L_X(q)-y\|_2^2.
$$

Como $y=L_X(p)$, tomando $q=p$, temos que

$$
\|L_X(p)-y\|_2^2
=
\|L_X(p)-L_X(p)\|_2^2
=
0.
$$

Assim, o menor erro possível é zero, já que uma norma nunca pode ser negativa.
Agora precisamos justificar que o polinômio minimizador será exatamente $p$. Como estamos supondo que

$$
m\geq d+1,
$$

pela questão 1(d), temos

$$
\operatorname{rank}(L_X)=d+1.
$$

Como $\dim(P_d)=d+1$, isso significa que $L_X$ é injetiva em $P_d$. Assim, se algum $q\in P_d$ também satisfaz

$$
\|L_X(q)-y\|_2^2=0,
$$

então

$$
L_X(q)=y=L_X(p).
$$

Pela injetividade de $L_X$, podemos concluir que

$$
q=p.
$$

Logo, o único polinômio de $P_d$ que minimiza

$$
\|L_X(q)-y\|_2^2
$$

é o próprio $p$. Portanto,

$$
R_{X,d}(L_X(p))=p.
$$

Como isso vale para todo $p\in P_d$, concluímos que

$$
R_{X,d}\circ L_X=\operatorname{id}_{P_d}.
$$

CURIOSIDADE:

Podemos tirar dessa questão que se os dados vieram de um polinômio $p$, a transformação $R_{x,d}$ vai recuperar exatamente esse $p$.


### b)

Se convença que $R_{X,d}$ é uma transformação linear.

O QUE EU QUERO QUE O BERNARDO LEIA:

Para mostrar que $R_{X,d}$ é linear, vamos escrever o problema em coordenadas na base canônica de $P_d$, de modo que a gente possa analisar $L_X$ como a matriz de Vandermonde.

Se $p\in P_d$, então

$$
p(x)=c_0+c_1x+\cdots+c_dx^d.
$$

Assim, podemos representar $p$ pelo vetor de coeficientes

$$
c=(c_0,c_1,\dots,c_d)^T\in\mathbb{R}^{d+1}.
$$

Na base canônica de $P_d$, temos

$$
L_X(p)=Vc,
$$

onde $V$ é a matriz de Vandermonde associada aos pontos $x_1,\dots,x_m$.

Pela definição de $R_{X,d}$, dado $b\in\mathbb{R}^m$, queremos encontrar o polinômio $p\in P_d$ que minimiza

$$
\|L_X(p)-b\|_2^2.
$$

Em coordenadas, isso é o mesmo que encontrar

$$
c_*=\operatorname{argmin}_{c\in\mathbb{R}^{d+1}}\|Vc-b\|_2^2.
$$

Esse é um problema de mínimos quadrados! Como $m\geq d+1$, pela questão 1(d)/(e), a matriz $V$ tem posto completo por colunas. Assim, $V^TV$ é invertível e a solução do problema de mínimos quadrados é único, sendo ela:

$$
c_*=V^+b,
$$

onde $V^+$ é a pseudoinversa de $V$. Logo, o vetor de coeficientes do polinômio minimizador $R_{X,d}(b)$ é

$$
V^+b.
$$

Como $V^+$ é uma matriz fixa, determinada só por $X$ e por $d$, a transformação

$$
b\mapsto V^+b
$$

é linear, isto é, para quaisquer $b_1,b_2\in\mathbb R^m$ e $\alpha,\beta\in\mathbb R$, temos

$$
R_{X,d}(\alpha b_1+\beta b_2)
=
\alpha R_{X,d}(b_1)+\beta R_{X,d}(b_2).
$$

Como cada vetor de coeficientes $c\in\mathbb{R}^{d+1}$ determina um único polinômio ( a identificação entre vetores de coeficientes em $\mathbb R^{d+1}$ e polinômios em $P_d$ é linear)

$$
c_0+c_1x+\cdots+c_dx^d
$$

podemos concluir que

$$
R_{X,d}:\mathbb{R}^m\to P_d
$$

é uma transformação linear.

### c)

Como calcular a matriz de $R_{X,d}$ na base canônica de $P_d$?


O QUE QUERO QUE BERNARDO LEIA:

Queremos mostrar a matriz da transformação $R_{x,d}$. Temos que cada coluna da matriz é a imagem de um vetor na base canônica do domínio.

Entretanto, a imagem desses vetores vai ser o polinômio minimizador associado a cada $e_i$ da base de $\mathbb{R}^m$. Porém, sabemos que é impossível termos uma matriz de polinômios.

Assim, vamos considerar $R$ uma transformação de $\mathbb{R}^m$ em $\mathbb{R}^{d+1}$, que associa cada vetor $e_i$ aos coeficientes do seu respectivo polinômio minimizador.

Quando escolhemos a base canônica de $P_d$, podemos identificar cada polinômio de $P_d$ com seu vetor de coeficientes em $\mathbb R^{d+1}$. Isso significa que estudar a matriz de $R_{X,d}$ é equivalente a estudar a transformação que envia cada $b\in\mathbb R^m$ ao vetor de coeficientes do polinômio minimizador. 

Logo, vamos definir a transformação $P$ que vai de $\mathbb{R}^m$ em $\mathbb{R}^m$, a projeção ortogonal de qualquer vetor sobre a imagem da representação matricial de $L_X(p)$, que agora vai estar no $\mathbb{R}^{d+1}$. Assim, para cada $b\in\mathbb R^m$, o vetor $Pb$ é o vetor de valores ajustados mais próximo de $b$, isto é, 

$$
Pb=Vc^*,
$$

onde

$$
c^*=\operatorname{argmin}_{c\in\mathbb R^{d+1}}\|Vc-b\|_2^2.
$$

Assim, a equação

$$
Pe_i = Vc_i^*
$$

vale para algum $c_i^*$, já que $Vc$ caracteriza aplicar os pontos $x_1,\dots,x_m$ no polinômio de coeficientes $c_i^*$.

A matriz $V$ é a representação matricial de $L_X(p)$, que, como $P_d$ está na base canônica, essa representação vai ser a matriz de Vandermonde.

Assim, temos que a matriz $V$ tem posto completo nas colunas, o que implica que podemos multiplicar o operador $V^+$, a pseudoinversa, nos dois lados, obtendo

$$
V^+Pe_i = c_i^*.
$$

onde $c_i^*$ é o vetor de coeficientes do polinôminio minimizador associado ao $e_i$. Como a matriz $V$ tem o posto completo nas colunas, a matriz de projeção ortogonal de vetores sobre a imagem será $VV^+$. Assim, temos que 

$$
V^+Pe_i = V^+VV^+e_i = I_{d+1}V^+e_i = V^+e_i = c_i^*.
$$

Como isso vale para todo $i=1,\dots,m$, a $i$-ésima coluna da matriz de $R_{X,d}$ é dada por

$$
V^+e_i.
$$

Portanto, a matriz inteira de $R_{X,d}$ é

$$
[R_{X,d}]
=
\begin{pmatrix}
| & | & & |\\
V^+e_1 & V^+e_2 & \cdots & V^+e_m\\
| & | & & |
\end{pmatrix}.
$$

Podemos fatorar $V^+$ à esquerda:

$$
[R_{X,d}]
=
V^+
\begin{pmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_m\\
| & | & & |
\end{pmatrix}.
[R_{X,d}]
=
V^+I_m
=
V^+.
$$

Assim, concluimos que:

$$
\boxed{
[R_{X,d}]=V^+
}
$$

Essa solução apenas reforça a ideia que vimos na letra anterior, explicando como funciona o processo de transformar a aplicação em outra equivalente que vai de $\mathbb{R}^m$ em $\mathbb{R}^{d+1}$. A conclusão que encontramos bate com os resultados que obtivemos até agora, já que as representações matriciais de $L_X$ e de $R_{x,d}$ quando compostas, serão $I_d$! (provamos isso na letra (a))



### d) Condicionamento

Escolha alguns valores de $m$ e $d$ e calcule os números de condicionamento das matrizes $L_X$ e $R_{X,d}$ na base canônica de $P_d$.

Como o número de condicionamento das matrizes $L_X$ e $R_{X,d}$ se comporta conforme $m$ e/ou $d$ aumentam?


Relembrando, temos que o número de condicionamento de uma matriz $A$ é dado por

$$
\kappa(A) = \frac{\sigma_{\max}}{\sigma_{\min}}
$$

onde $\sigma_{\max}$ e $\sigma_{\min}$ são o maior e menor valor singular, respectivamente.
Temos que se $V$ tem a seguinte fatoração SVD reduzida:

$$
V
=
\hat{U}\hat{\Sigma}\hat{W^T}
$$

Então, temos que a pseudoinversa de V será:

$$
V^+
=
\hat{W}\hat{\Sigma^+}\hat{U^T}
$$

onde:

$$
\hat{\Sigma^+}

=
\operatorname{diag}
\left(
\frac{1}{\sigma_0},
\frac{1}{\sigma_{1}},
\dots,
\frac{1}{\sigma_d}
\right).
$$

Porém, saber dessa fatoração SVD reduzida de $V^+$ implica que:

$$
\kappa(V) = \kappa(V^+)
$$

Pois:

$$
\kappa(V^+)
=
\frac{\frac{1}{\sigma_d}}{\frac{1}{\sigma_0}}
=
\frac{\sigma_0}{\sigma_d}
=
\kappa(V).
$$


Na próxima célula, vamos verificar se o condicionamento das duas matrizes são computacionalmente equivalentes como provamos teoricamente. 

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors

# Fixamos m para comparar os métodos variando d
m = 70
X = np.linspace(-1, 1, m)
conds_V = []
conds_pseudoV = []
conds_pseudo_np_V = []
valores_d = list(range(1, m))  # d = 1, ..., m-1

for d in valores_d:
    V = np.vander(X, N=d+1, increasing=True)

    sigma_V = np.linalg.svd(V, compute_uv=False)
    k_V = sigma_V[0] / sigma_V[-1]
    k_pseudo_V = (1 / sigma_V[-1]) / (1 / sigma_V[0])
    pseudo_V = np.linalg.pinv(V)

    conds_V.append(k_V)
    conds_pseudoV.append(k_pseudo_V)
    conds_pseudo_np_V.append(np.linalg.cond(pseudo_V))



plt.figure(figsize=(9, 6))

plt.plot(
    valores_d,
    conds_V,
    label="cond V",
    marker="o",
    markersize=3
)

plt.plot(
    valores_d,
    conds_pseudoV,
    label="cond V⁺ teórica",
    linestyle="--"
)

plt.plot(
    valores_d,
    conds_pseudo_np_V,
    label="cond np.linalg.pinv(V)",
    linestyle=":",
    marker="x",
    markersize=3
)

plt.yscale("log")

plt.xlabel("Grau do polinômio d")
plt.ylabel("Condicionamento")
plt.title(f"Comparação entre $\\kappa(V)$, $\\kappa(V^+)$ teórico e $\\kappa(\\mathrm{{pinv}}(V))$ para m={m}")

plt.legend()
plt.grid(True, which="both", alpha=0.3)

plt.show()
```

```python
valores_m = list(range(2, 100))
valores_d = list(range(1, 70))

grade_conds = np.full((len(valores_m), len(valores_d)), np.nan)

for idx_m, m in enumerate(valores_m):
    for idx_d, d in enumerate(valores_d):

        # A questão supõe m >= d + 1
        if m < d + 1:
            continue

        X = np.linspace(-1, 1, m)

        # P_d tem dimensão d+1: colunas 1, x, x^2, ..., x^d
        V = np.vander(X, N=d+1, increasing=True)

        try:
            grade_conds[idx_m, idx_d] = np.linalg.cond(V)
        except np.linalg.LinAlgError:
            grade_conds[idx_m, idx_d] = np.inf


# Para o mapa de calor ficar legível, usamos escala logarítmica.
# Também cortamos valores infinitos para um teto visual.
grade_plot = np.copy(grade_conds)
grade_plot[np.isinf(grade_plot)] = 1e16

plt.figure(figsize=(10, 8))

im = plt.imshow(
    grade_plot,
    extent=[valores_d[0], valores_d[-1], valores_m[0], valores_m[-1]],
    origin="lower",
    cmap="viridis",
    norm=colors.LogNorm(vmin=1, vmax=1e16),
    aspect="auto",
)

plt.colorbar(im, label="Condicionamento da matriz")
plt.xlabel("Grau do polinômio d")
plt.ylabel("Número de pontos m")
plt.title("Condicionamento da matriz de $L_X$ na base canônica")

plt.show()
```

### e)

Calcule as matrizes de $L_X$ e $R_{X,d}$ usando a base dos polinômios interpoladores de Lagrange.

Qual o condicionamento das matrizes $L_X$ e $R_{X,d}$ nestas bases?


O QUE QUERO QUE O BERNARDO LEIA:

Vamos analisar os dois casos das dimensões do domínio e contradomínio.

### 1) $m = d+1$.

Relembrando, temos que cada coluna da matriz de uma transformação linear é a imagem de um vetor da base do domínio. Assim, para calcular a matriz de $L_X$ na base dos polinômios interpoladores de Lagrange, basta calcular $L_X(p_i)$ para cada $i$.

Pela definição dos polinômios interpoladores de Lagrange, temos

$$
p_i(x_j)=\delta_{ij}.
$$

Logo,

$$
L_X(p_i)
=
\bigl(p_i(x_1),p_i(x_2),\dots,p_i(x_m)\bigr).
$$

No caso $m=d+1$, vamos ter

$$
L_X(p_i)
=
(0,\dots,0,1,0,\dots,0)
=
e_i,
$$

onde o $1$ aparece na $i$-ésima posição.

Assim, a matriz de $L_X$ na base dos polinômios interpoladores de Lagrange é

$$
\boxed{
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
1 & 0 & \cdots & 0\\
0 & 1 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 1
\end{pmatrix}
=
I_{d+1}.
}
$$

Relembrando, temos que o condicionamento de uma matriz inversível $A$ é dado por

$$
\kappa(A)=\|A\|\|A^{-1}\|.
$$

Como a gente viu, na base dos polinômios interpoladores de Lagrange, a matriz de $L_X$ é a própria identidade:

$$
[L_X]_{\mathcal{B}_{Lag}}=I.
$$

Assim,

$$
\kappa(I)=\|I\|\|I^{-1}\| =\|I\|^2.
$$

Agora vamos calcular $\|I\|$. Pela definição de norma induzida,
$$
\|I\|
=
\sup_{x\neq 0}
\frac{\|Ix\|}{\|x\|}
=
\sup_{x\neq 0}
\frac{\|x\|}{\|x\|}
=
1.
$$

Portanto,

$$
\kappa(I)
=
\|I\|^2
=
1^2
=
1.
$$

Assim,

$$
\kappa\left([L_X]_{\mathcal{B}_{Lag}}\right)=1.
$$

No caso da matriz de transformação de $R_{x,d}$, temos que encontrar o polinômio minimizador na base dos polinômios intepoladores de Lagrange. Vamos então aplicar a $R_{x,d}$ na base canônica do $\mathbb{R}^m$. Assim,

$$
R_{X,d}(e_i)
=
\operatorname{argmin}_{q\in P_d}
\|L_X(q)-e_i\|_2^2.
$$

Pela construção dos polinômios interpolares de Lagrange, vimos que:

$$
L_X(p_i)
=
(0,\dots,0,1,0,\dots,0)
=
e_i,
$$

Logo, temos que o polinômio minimizador $ R_{X,d}(e_i) $ será o próprio $p_i$. Mas, na base de Lagrange, o vetor de coordenadas de $p_i$ vai ser justamente $e_i$. Assim, como isso vale para todos os vetores da base canônica do $\mathbb{R}^m$, temos que:

$$
\boxed{
[R_{x,d}]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_m\\
| & | & & |
\end{pmatrix}
=
I_{d+1}
}
$$

Como vimos,  

$$
\kappa(I)=1.
$$

Logo, concluimos que quando $m = d+1$:

$$
\boxed{
\kappa\left([R_{x,d}]_{\mathcal{B}_{Lag}}\right)= \kappa[L_X]_{\mathcal{B}_{Lag}}
= 1.
}
$$


### 2) $ m>d+1 $:

Quando $m > d + 1$, temos mais pontos no conjunto $X$ = {${x_1, x_2, x_3, ... , x_m}$} do que polinômios na base de Lagrange. Não é possível definir uma base de $P_d$ com os polinômios interpoladores de Lagrange, já que dizer que:

$$
p_i(x_j)=\delta_{ij}.
$$

para $j = 1, 2, ..., m$ quando $m > d + 1$, estaremos impondo mais condições do que coeficientes disponíveis (temos até d+1 coeficientes). Assim, vamos tomar {${x_1, x_2, x_3, ... , x_{d+1}}$} pontos de $X$ e construir o conjunto na base de Lagrange {${p_1, p_2, p_3, ... , p_{d+1}}$} tal que:

$$
p_i(x_j)=\delta_{ij}.
$$

para $j = 1, 2, ..., d+1$. Assim, teremos uma base de Lagrange restrita à condição da dimensão do espaço $P_d$. Entretanto, como $L_x$ vai avaliar $p$ em todos os $m$ pontos de $X$, temos que:

$$
L_X(p_i)
=
\begin{pmatrix}
p_i(x_1)\\
\vdots\\
p_i(x_i)\\
\vdots\\
p_i(x_{d+1})\\
p_i(x_{d+2})\\
\vdots\\
p_i(x_m)
\end{pmatrix}.
$$

Então a coluna $i$ da representação matricial de $L_X$ é:

$$
L_X(p_i)
=
\begin{pmatrix}
0\\
\vdots\\
1\\
\vdots\\
0\\
p_i(x_{d+2})\\
\vdots\\
p_i(x_m)
\end{pmatrix}.
$$

Para que possamos calcular a matriz de $L_X$, precisamos saber quem são os polinômios na base de Lagrange aplicados em $x_{d+2},\dots,x_m$. Mas, temos uma fórmula fechada pra isso, que é:

$$

p_i(x)
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x-x_j}{x_i-x_j},
\qquad i=1,\dots,d+1.
$$

Portanto, a matriz de $L_X$ na base de Lagrange tem entradas

$$
L_{ki}
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x_k-x_j}{x_i-x_j}.
$$

Logo,

$$
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix},
$$

onde o bloco $*$ contém os valores

$$
p_i(x_k),
\qquad
k=d+2,\dots,m,
\quad
i=1,\dots,d+1.
$$

Vamos agora encontrar a representação matricial de $R_{x,d}$ quando $P_d$ está na base dos polinômios interpoladores de Lagrange. Para cada elemento da base canônica do $\mathbb{R}^m$, vamos encontrar $R_{x,d}(e_i)$. Nós vimos que a forma matricial de $R_{x,d}$ será a pseudoinversa de A:

$$
[R_{x,d}]_{\mathrm{Lag}}
=
A^+
$$

onde A é a representação matricial de $L_X$ se a matriz de $L_X$ tem posto completo nas colunas. Como a matriz de $L_X$ tem posto completo nas colunas, (um de seus blocos é a identidade $d+1$ x $d+1$) podemos afirmar que:

$$
\boxed{
[R_{x,d}]_{\mathrm{Lag}}
=
A^+
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix}
^+
}
$$

onde:

$$
A^+=(A^TA)^{-1}A^T.
$$

Porém, como calculamos os condicionamentos de $[R_{x,d}]_{\mathrm{Lag}}$ e de $[L_X]_{\mathrm{Lag}}$? Como vimos na letra d), podemos utilizar o fato de que:

$$
\kappa(A) = \frac{\sigma_{\max}}{\sigma_{\min}}
$$

onde $\sigma_{\max}$ e $\sigma_{\min}$ são o maior e menor valor singular, respectivamente.

CURIOSIDADES:

O condicionamento da Rx ser 1 é muito legal, porque o condicionamento da matriz mede a maior distorção que a matriz pode causar nos vetores do domínio, então o condicionamento da representação matricial de $R_{x,d}$ ser 1 significa que cada coordenada da entrada dos dados vai parar diretamente na coordenada correspondente da saída, então a matriz não vai amplificar erros relativos!! 

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors


def lagrange_basis_eval(a, nodes, i):
    """
    Calcula p_i(a), onde p_i é o polinômio de Lagrange
    associado aos pontos em 'nodes'.

    Aqui i começa em 0.
    """
    produto = 1.0
    xi = nodes[i]

    for j, xj in enumerate(nodes):
        if j != i:
            produto *= (a - xj) / (xi - xj)

    return produto


def matriz_L_lagrange(X, d):
    """
    Matriz de L_X na base de Lagrange de P_d.

    A base de Lagrange é construída usando d+1 pontos
    espalhados dentro de X.

    Depois avaliamos esses polinômios em todos os m pontos de X.

    Dimensão: m x (d+1).
    """
    X = np.asarray(X)

    if len(X) < d + 1:
        raise ValueError("Precisamos de m >= d+1.")

    # Escolhe d+1 pontos espalhados dentro de X
    indices = np.linspace(0, len(X) - 1, d + 1, dtype=int)
    nodes = X[indices]

    m = len(X)
    A = np.zeros((m, d + 1))

    for k in range(m):
        for i in range(d + 1):
            A[k, i] = lagrange_basis_eval(X[k], nodes, i)

    return A


valores_m = list(range(2, 300))
valores_d = list(range(1, 300))

grade_conds = np.full((len(valores_m), len(valores_d)), np.nan)

for idx_m, m in enumerate(valores_m):
    for idx_d, d in enumerate(valores_d):

        # A questão supõe m >= d + 1
        if m < d + 1:
            continue

        X = np.linspace(-1, 1, m)

        A = matriz_L_lagrange(X, d)

        try:
            grade_conds[idx_m, idx_d] = np.linalg.cond(A)
        except np.linalg.LinAlgError:
            grade_conds[idx_m, idx_d] = np.inf


# Para o mapa de calor ficar legível, usamos escala logarítmica.
grade_plot = np.copy(grade_conds)
grade_plot[np.isinf(grade_plot)] = 1e16

plt.figure(figsize=(10, 8))

im = plt.imshow(
    grade_plot,
    extent=[valores_d[0], valores_d[-1], valores_m[0], valores_m[-1]],
    origin="lower",
    cmap="viridis",
    norm=colors.LogNorm(vmin=1, vmax=1e16),
    aspect="auto",
)

plt.colorbar(im, label="Condicionamento da matriz")
plt.xlabel("Grau do polinômio d")
plt.ylabel("Número de pontos m")
plt.title("Condicionamento da matriz de $L_X$ na base de Lagrange")

plt.show()
```

Compare a complexidade computacional de calcular $L_X$ e $R_{X,d}$ usando as bases canônica e de Lagrange com o condicionamento destas matrizes.

---

Vamos analisar a complexidade computacional de calcular $L_X$ e $R_{x,d}$ em cada uma das bases:

### 1) Base canônica:

Analisando a matriz de $L_X$:

Temos que a matriz de $L_X$ é $V$, onde a matriz V é a matriz de Vandermonde associada aos pontos de $X$. Note que na $i$-ésima linha de V, estamos calculando potências de $x_i$ de grau até $d$. Assim, em cada linha, a complexidade é de até $O(d)$. Como isso vale para todas as $m$ linhas da matriz V, temos que a complexidade computacional de calcular $L_X$ é:

$$
\boxed{
L_X \rightarrow
O(md)
}
$$

Assim, temos que se $m \approx d$, a complexidade se torna de ordem quadrática, o que nunca é bom.

Analisando a matriz de $R_{x,d}$:

Na matriz de $R_{x,d}$, temos que calcular a pseudoinversa da matriz de Vandermonde $V$. Podemos prosseguir de três formas diferentes:

**1. Por meio da fatoração reduzida SVD de $V$:**

A complexidade computacional de calcular a fatoração SVD reduzida no NumPy (fazendo a análise assintótica apenas) de uma matriz $m\times(d+1)$ quando $m$ $\geq$ $d+1$, é de 

$$
O(m(d+1)^2)
$$ 

Sabendo a fatoração reduzida de V:

$$
V
=
\hat{U}\hat{\Sigma} \hat{W}^T
$$

A pseudoinversa de V será:

$$
V^+
=
\hat{W}\hat{\Sigma^+} \hat{U}^T
$$

Como encontramos a SVD reduzida de $V$, temos que:

$$
\hat{\Sigma}
=
\begin{pmatrix}
\operatorname{diag}(\sigma_0,\sigma_1,\dots,\sigma_{d})\\
\end{pmatrix}.
$$

A sua pseudoinversa será:

$$
\hat{\Sigma^+}

=
\operatorname{diag}
\left(
\frac{1}{\sigma_{0}},
\frac{1}{\sigma_{1}},
\dots,
\frac{1}{\sigma_d}
\right).
$$


Calcular essa pseudoinversa tem a complexidade computacional de inverter cada entrada dos elementos da diagonal (custo $\approx$ $O(d+1)$), que tem ordem menor que a complexidade de calcular a SVD reduzida de $V$. Assim, vimos que a etapa mais custosa vai ser de calcular a fatoração SVD reduzida de $V$, com custo 

$$
O(m(d+1)^2)
$$

Se quisermos explicitar a matriz $V^+$ e calcular o produto das matrizes, a complexidade computacional fica na mesma ordem de $O(m(d+1)^2)$, ou até menor. 

Por fim, a complexidade computacional de calcular a $R_{x,d}$ utilizando a fatoração SVD reduzida de $V$ é:

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$


**2. Por meio da fatoração QR reduzida de $V$:**

Sabemos que, se a matriz $V$ tem fatoração QR reduzida:

$$
V=\hat Q\hat R,
$$

Como visto em aula, a pseudoinversa de $V$ será (se a matriz $V$ tem posto completo, nosso caso):

$$

V^+
=
\hat R^{-1}\hat Q^T.
$$

Agora vamos analisar a complexidade computacional. A fatoração QR reduzida de uma matriz $V$ $m\times(d+1)$, quando $m \geq d+1$ tem complexidade computacional da ordem de

$$
O(m(d+1)^2).
$$

Se quisermos formar explicitamente a matriz

$$
V^+=\hat R^{-1}\hat Q^T,
$$

não precisamos calcular $\hat R^{-1}$ diretamente. Podemos resolver o sistema triangular

$$
\hat R X=\hat Q^T.
$$

Como $\hat R$ tem dimensão $(d+1)\times(d+1)$ e $\hat Q^T$ tem dimensão $(d+1)\times m$, resolver esse sistema triangular com $m$ lados direitos tem complexidade da ordem de

$$
O(m(d+1)^2).
$$

Assim, a etapa de formar explicitamente $V^+$ também fica na ordem de

$$
O(m(d+1)^2).
$$

Logo, a complexidade computacional para calcular $R_{X,d}$ usando a fatoração QR reduzida de $V$ fica na ordem de

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$

Assintoticamente, QR reduzida e SVD reduzida ficam na mesma ordem $O(m(d+1)$ quando $m \geq d+1$. Porém, QR costuma ser computacionalmente mais barata (brincadeiras com matrizes ortogonais), enquanto SVD é mais informativa e mais robusta para analisar posto, valores singulares e condicionamento.


**3. Calcular o produto $V^+ = (V^TV)^{-1}V^T$;**

Note a priori que se a matriz $V$ é mal condicionada, esse método é uma péssima escolha, já que:

$$
\kappa(V^TV)
\approx
\kappa(V)^2
$$

Então, se o condicionamento de $V$ é por exemplo $10^6$, então o condicionamento de $V^TV$ será $10^{12}$. (explicar melhor)

A complexidade computacional de calcular $V^TV$ é (produto matricial)

$$
O(m(d+1)^2)
$$

A partir desse produto, temos dois caminhos:


i) Resolver o sistema:

$$
(V^TV)X
=
V^T
$$

ii) Resolver a equação:

$$
x
=
(V^TV)^{-1}V^T
$$

Ao resolver o sistema (i) por fatoração (LU), temos que a complexidade é de 

$$
O((d+1)^3)
$$

e a substituição vai ter complexidade:

$$
O(m(d+1)^2)
$$

Assim, a complexidade final será:

$$
O(m(d+1)^2 + m(d+1) + (d+1)^3 + m(d+1)^2) \approx O(m(d+1)^2 + (d+1)^3)
$$

Ou seja (já que $m \geq d+1$),

$$
O(m(d+1)^2)
$$


Resolvendo a equação em (ii), obteríamos que calcular a inversa ($V^TV$ tem dimensões $(d+1)\times{(d+1)}$) tem complexidade computacional:

$$
O((d+1)^3)
$$

Depois, fazer o produto de $(V^TV)^{-1}$ com ${V^T}$ tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Assim, a complexidade computacional é a mesma:

$$
O(m(d+1)^2 + (d+1)^3 + m(d+1)^2) =  O(m(d+1)^2 + (d+1)^3)
$$

isto é,

$$
O(m(d+1)^2)
$$

Assim, de ambos os métodos, podemos concluir que a complexidade computacional de calcular a matriz de $R_{x,d}$ por meio da fórmula fechada da pseudoinversa é:

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$

Assim, assintoticamente falando, temos que a complexidade computacional de calcular a matriz $R_{x,d}$ é de 

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$

OBS: Apesar de assintoticamente todos os métodos serem iguais, temos que cada um acompanha uma constante diferente e cada um deles garantem propriedades diferentes. Por exemplo, a SVD é a fatoração mais informativa, porém a fatoração QR chega a ser a mais barata dentre todas. Já o método do produto nos faz calcular a matriz quadrada $V^TV$, que o condicionamento é aprox. o quadrado do condicionamento de $V$.

### 2) Base dos Polinômios Interpoladores de Lagrange.

Na base dos polinômios interpoladores de Lagrange, precisamos analisar os dois casos das dimensões do domínio e contradomínio:

**i) $m$ = $d+1$**

Na base de Lagrange, temos que a matriz de $L_X$ e de $R_{x,d}$ são do formato:

$$
[R_{x,d}]_{\mathcal{B}_{Lag}}= [L_X]_{\mathcal{B}_{Lag}}
= I_{d+1}
$$

Isso implica que a complexidade computacional de calcular as matrizes de $L_X$ e $R_{x,d}$ é:

$$
\boxed{
R_{x,d}, L_X \rightarrow
O(m(d+1))
}
$$


**ii) $m > d+1$:**

Vimos que a matriz de $L_X$ quando $m > d+1$:

$$
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix},
$$

e a matriz de $R_{x,d}$ será:

$$
[R_{x,d}]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix} ^+
$$

onde o bloco $*$ contém os valores

$$
p_i(x_k),
\qquad
k=d+2,\dots,m,
\quad
i=1,\dots,d+1.
$$


A complexidade computacional de calcular cada entrada é a complexidade de calcular o produtório:

$$
p_i(x_k)
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x_k-x_j}{x_i-x_j},
\qquad i=1,\dots,d+1.
$$

que tem complexidade:

$$
O(d)
$$

Temos que o bloco $*$ tem $m-(d+1)$ linhas e $d+1$ colunas, o que nos diz que a complexidade de calcular o bloco $*$ terá complexidade computacional:

$$
O(d(m-d-1)(d+1))
$$

Assintoticamente, quando $m > d+1$:

$$
O(d(m-d-1)(d+1)) = O(m(d+1)^2)
$$

Calcular o bloco identidade $I^{d+1}$ tem complexidade $O((d+1)^2)$. Assim, a complexidade computacional total de calcular a matriz $L_X$ é:

$$
\boxed{
L_X \rightarrow
O(m(d+1)^2)
}
$$

Vamos analisar a pseudoinversa da matriz de $L_X$, isto é, a matriz de $R_{x,d}$. 

Para calcularmos a matriz de $R_{x,d}$, precisamos primeiro montar a matriz de $L_X$ e depois calcular a sua pseudoinversa. Vimos que montar a matriz de $L_X$ tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Provamos anteriormente que a complexidade computacional de calcular $A^+$ (quando A tem posto completo) pela fatoração QR, SVD e equações normais tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Assim, como a matriz de $L_X$ tem posto completo, temos que:

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$


Vamos relacionar os condicionamentos dessas matrizes:
