## Questão 2. Regressão polinomial

Suponha que $m\geq d+1$.

Defina a função

$$
R_{X,d}:\mathbb{R}^m\to P_d
$$

que associa a cada vetor $b\in\mathbb{R}^m$ o polinômio $p\in P_d$ que melhor se ajusta aos valores $b_i$ nos pontos $x_i\in X$ no sentido de mínimos quadrados, ou seja,

$$
R_{X,d}(b)
=
\operatorname{argmin}_{p\in P_d}
\|L_X(p)-b\|_2^2.
$$



### a)

Mostre que $R_{X,d}$ é uma inversa à esquerda de $L_X$, ou seja,

$$
R_{X,d}\circ L_X=\operatorname{id}_{P_d}.
$$



Queremos mostrar que:

$$
R_{X,d}\circ L_X = \operatorname{id}_{P_d}.
$$

Ou seja, para todo polinômio $p\in P_d$, queremos provar que

$$
R_{X,d}(L_X(p))=p.
$$

Antes de qualquer coisa, precisamos saber se $R_{x,d}$ está bem definida. Porém, isso decorre da unicidade da $\operatorname{argmin}$! Então vamos prosseguir:

Seja $p\in P_d$. Aplicando $L_X$ a $p$, vamos obter o vetor

$$
L_X(p)
=
(p(x_1),p(x_2),\dots,p(x_m)).
$$

Vamos chamar esse vetor de

$$
y=L_X(p)\in \mathbb{R}^m.
$$

Pela definição de $R_{X,d}$, temos

$$
R_{X,d}(y)
=
\operatorname{argmin}_{q\in P_d}
\|L_X(q)-y\|_2^2.
$$

Como $y=L_X(p)$, tomando $q=p$, temos que

$$
\|L_X(p)-y\|_2^2
=
\|L_X(p)-L_X(p)\|_2^2
=
0.
$$

Assim, o menor erro possível é zero, já que uma norma nunca pode ser negativa.
Agora precisamos justificar que o polinômio minimizador será exatamente $p$. Como estamos supondo que

$$
m\geq d+1,
$$

pela questão 1(d), temos

$$
\operatorname{rank}(L_X)=d+1.
$$

Como $\dim(P_d)=d+1$, isso significa que $L_X$ é injetiva em $P_d$. Assim, se algum $q\in P_d$ também satisfaz

$$
\|L_X(q)-y\|_2^2=0,
$$

então

$$
L_X(q)=y=L_X(p).
$$

Pela injetividade de $L_X$, podemos concluir que

$$
q=p.
$$

Logo, o único polinômio de $P_d$ que minimiza

$$
\|L_X(q)-y\|_2^2
$$

é o próprio $p$. Portanto,

$$
R_{X,d}(L_X(p))=p.
$$

Como isso vale para todo $p\in P_d$, concluímos que

$$
R_{X,d}\circ L_X=\operatorname{id}_{P_d}.
$$

Comentários:

Podemos tirar dessa questão que se os dados vieram de um polinômio $p$, a transformação $R_{x,d}$ vai recuperar exatamente esse $p$.


### b)

Se convença que $R_{X,d}$ é uma transformação linear.


Para mostrar que $R_{X,d}$ é linear, vamos escrever o problema em coordenadas na base canônica de $P_d$, de modo que a gente possa analisar $L_X$ como a matriz de Vandermonde.

Se $p\in P_d$, então

$$
p(x)=c_0+c_1x+\cdots+c_dx^d.
$$

Assim, podemos representar $p$ pelo vetor de coeficientes

$$
c=(c_0,c_1,\dots,c_d)^T\in\mathbb{R}^{d+1}.
$$

Na base canônica de $P_d$, temos

$$
L_X(p)=Vc,
$$

onde $V$ é a matriz de Vandermonde associada aos pontos $x_1,\dots,x_m$.

Pela definição de $R_{X,d}$, dado $b\in\mathbb{R}^m$, queremos encontrar o polinômio $p\in P_d$ que minimiza

$$
\|L_X(p)-b\|_2^2.
$$

Em coordenadas, isso é o mesmo que encontrar

$$
c_*=\operatorname{argmin}_{c\in\mathbb{R}^{d+1}}\|Vc-b\|_2^2.
$$

Esse é um problema de mínimos quadrados! Como $m\geq d+1$, pela questão 1(d)/(e), a matriz $V$ tem posto completo por colunas. Assim, $V^TV$ é invertível e a solução do problema de mínimos quadrados é única, sendo ela:

$$
c_*=V^+b,
$$

onde $V^+$ é a pseudoinversa de $V$. Logo, o vetor de coeficientes do polinômio minimizador $R_{X,d}(b)$ é

$$
V^+b.
$$

Como $V^+$ é uma matriz fixa, determinada só por $X$ e por $d$, a transformação

$$
b\mapsto V^+b
$$

é linear, isto é, para quaisquer $b_1,b_2\in\mathbb R^m$ e $\alpha,\beta\in\mathbb R$, temos

$$
R_{X,d}(\alpha b_1+\beta b_2)
=
\alpha R_{X,d}(b_1)+\beta R_{X,d}(b_2).
$$

Assim, podemos concluir que

$$
R_{X,d}:\mathbb{R}^m\to P_d
$$

é uma transformação linear. 



### c)

Como calcular a matriz de $R_{X,d}$ na base canônica de $P_d$?


Temos que cada coluna da matriz é a imagem de um vetor na base canônica do domínio. A imagem desses vetores vai ser o polinômio minimizador associado a cada $e_i$ da base de $\mathbb{R}^m$. Porém, sabemos que é impossível termos uma matriz de polinômios.

Assim, como no item (b), vamos considerar $R$ a transformação de $\mathbb{R}^m$ em $\mathbb{R}^{d+1}$, que associa cada vetor $e_i$ aos coeficientes do seu respectivo polinômio minimizador.

Pelo resultado do item (b), temos que aplicar a transformação $R_{x,d}$ ao vetor $e_i$ da base canônica do $\mathbb{R}^m$ nos dá:

$$
c_i^* = V^+e_i.
$$

Como isso vale para todo $i=1,\dots,m$, a matriz inteira de $R_{X,d}$ é

$$
[R_{X,d}]
=
\begin{pmatrix}
| & | & & | & | \\
c_1^* & c_2^* & c_3^* & \cdots & c_m^*\\
| & | & & | & |
\end{pmatrix}
=
\begin{pmatrix}
| & | & & |\\
V^+e_1 & V^+e_2 & \cdots & V^+e_m\\
| & | & & |
\end{pmatrix}.
$$

Podemos fatorar $V^+$ a esquerda:

$$
[R_{X,d}]
=
V^+
\begin{pmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_m\\
| & | & & |
\end{pmatrix}
=
V^+I_m
=
V^+.
$$

Assim, podemos concluir que:

$$
\boxed{
[R_{X,d}]=V^+
}
$$




### d) Condicionamento

Escolha alguns valores de $m$ e $d$ e calcule os números de condicionamento das matrizes $L_X$ e $R_{X,d}$ na base canônica de $P_d$.

Como o número de condicionamento das matrizes $L_X$ e $R_{X,d}$ se comporta conforme $m$ e/ou $d$ aumentam?


Relembrando, temos que o número de condicionamento de uma matriz $A$ é dado por

$$
\kappa(A) = \frac{\sigma_{\max}}{\sigma_{\min}}
$$

onde $\sigma_{\max}$ e $\sigma_{\min}$ são o maior e menor valor singular, respectivamente.
Temos que se $V$ tem a seguinte fatoração SVD reduzida:

$$
V
=
\hat{U}\hat{\Sigma}\hat{W}^T
$$

Então, temos que a pseudoinversa de V será:

$$
V^+
=
\hat{W}\hat{\Sigma}^{-1}\hat{U}^T
$$

onde:

$$
\hat{\Sigma}^{-1}

=
\operatorname{diag}
\left(
\frac{1}{\sigma_0},
\frac{1}{\sigma_{1}},
\dots,
\frac{1}{\sigma_d}
\right).
$$

Porém, saber dessa fatoração SVD reduzida de $V^+$ implica que:

$$
\kappa(V) = \kappa(V^+)
$$

Pois:

$$
\kappa(V^+)
=
\frac{\frac{1}{\sigma_d}}{\frac{1}{\sigma_0}}
=
\frac{\sigma_0}{\sigma_d}
=
\kappa(V).
$$


Na próxima célula, vamos analisar se isso é verdade numericamente para diferentes tipos de implemetação da pseudoinversa em Python.

```python

def cond_svd(A, tol=1e-18):
    sigma = np.linalg.svd(A, compute_uv=False)

    if sigma[-1] < tol:
        return np.inf

    return sigma[0] / sigma[-1]

m = 150
valores_d = list(range(1, 80))

X = np.linspace(-1, 1, m)

ds = []
conds_V = []
conds_Rx_pinv = []
conds_Rx_normal = []


for d in valores_d:

    if m < d + 1:
        continue

    V = np.vander(X, N=d+1, increasing=True)

    R_x_pinv = np.linalg.pinv(V)

    V_trans = V.T
    R_x = np.linalg.inv(V_trans @ V) @ V_trans

    cond_V = np.linalg.cond(V)
    cond_Rx_pinv = np.linalg.cond(R_x_pinv)
    cond_Rx = np.linalg.cond(R_x)

    ds.append(d)
    conds_V.append(cond_V)
    conds_Rx_pinv.append(cond_Rx_pinv)
    conds_Rx_normal.append(cond_Rx)

plt.figure(figsize=(9, 6))

plt.plot(ds, conds_V, label=r"$\kappa(V)$")
plt.plot(ds, conds_Rx_pinv, label=r"$\kappa(\mathrm{pinv}(V))$")
plt.plot(ds, conds_Rx_normal, label=r"$\kappa((V^T V)^{-1}V^T)$")

plt.yscale("log")

plt.xlabel("d")
plt.ylabel("condicionamento")
plt.title("cond dos 3 métodos em função do grau d")

plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()

plt.show()

```

Nesse gráfico, estamos fixando $m = 150$ (a variação do grau $d$ teve efeito muito mais forte no condicionamento do que a variação de $m$, mas vamos analisar isso melhor no próximo item) e analisando o condicionamento das 3 diferentes matrizes na base canônica: $V$, $V^+$ calculado por meio do np.linalg.pinv e $V^+$ calculado utlizando o produto das matrizes que compõem a pseudoinversa de $V$.

Note que o np.linalg.pinv usa SVD internamente com tolerância, então ela continua produzindo uma pseudoinversa mesmo quando o problema já tá numericamente muito instável. 

Concluímos que o condicionamento de $V$ e de $V^+$ são iguais teoricamente, mas numericamente isso não acontece sempre, pois a partir do momento que $V$ está muito mal condicionada, a máquina não consegue mais distinguir o menor valor singular de $V$ de zero. O gráfico nos mostra que cond($V$) $\approx$ cond($V^+$) para quando os valores do condicionamento são razoáveis ($\approx$ $d \leq 43$). Quando o menor valor singular de $V$ está muito próximo do erro de precisão da máquina, a função np.linalg.pinv aproxima o mesmo de zero, o que faz com que não possamos confiar no resultado do condicionamento dos dois métodos. 

Note que a priori, cond($V^TV$) $\approx$ cond($V)^2$. Assim, o condicionamento do cálculo da pseudoinversa cresce bastante, mas fica abaixo do cond($V$) para graus maiores. Isso não quer dizer que ela é mais estável, já que como estamos formando $V^TV$, o condicionamento piora aproximadamente com o quadrado do condicionamento de $V$.

Então a matriz calculada pela np.linalg.inv pode estar sendo afetada por erro numérico e cancelamento, o que nos diz que ela não é uma boa referência de estabilidade, porque ela já tá "contaminada" pelo mau condicionamento de $V^TV$. Assim, ainda que a curva vai parecer menor em alguns trechos, mas o procedimento de calcular com a inversa é numericamente menos confiável do que calcular pelos outros métodos.

Podemos concluir que NUMERICAMENTE, cond($V$) $\approx$ cond($V^+$) para valores confiáveis do condicionamento de $V$. Teoricamente, isso é uma propriedade matemática que provamos, mas quando a matriz já está mal condicionada, os erros numéricos são mais fortes do que a equivalência entre os condicionamentos.

OBS:
Assim, daqui pra frente, as análises feitas sobre o condicionamento da matriz de $L_X$ será equivalente às análises das matrizes de $R_{x,d}$, salvo os casos onde a matriz já está tão mal condicionada que não conseguimos abstrair nada dos resultados de $L_x$, e assim, também da matriz de $R_{x,d}$.


```python
valores_m = list(range(2, 150))
valores_d = list(range(1, 80))

grade_conds = np.full((len(valores_m), len(valores_d)), np.nan)


def cond_heatmap_canonica(x_menor, x_maior):
    for idx_m, m in enumerate(valores_m):

        X = np.linspace(x_menor, x_maior, m)

        for idx_d, d in enumerate(valores_d):

            if m < d + 1:
                continue

            V = np.vander(X, N=d+1, increasing=True)

            try:
                grade_conds[idx_m, idx_d] = cond_svd(V)
            except np.linalg.LinAlgError:
                grade_conds[idx_m, idx_d] = np.inf


    grade_plot = np.copy(grade_conds)
    grade_plot[np.isinf(grade_plot)] = 1e16

    plt.figure(figsize=(10, 8))

    im = plt.imshow(
        grade_plot,
        extent=[valores_d[0], valores_d[-1], valores_m[0], valores_m[-1]],
        origin="lower",
        cmap="viridis",
        norm=colors.LogNorm(vmin=1, vmax=1e20),
        aspect="auto",
    )

    plt.colorbar(im, label="$cond(L_X)$")
    plt.xlabel("grau do polinômio d")
    plt.ylabel("número de pontos m")
    plt.title(f"cond($L_X$) = cond($R_xd$) na base canônica em linspace({x_menor}, {x_maior}, m)")

    plt.show()
```

```python
cond_heatmap_canonica(0, 1)
cond_heatmap_canonica(-1, 0)
cond_heatmap_canonica(-1, 1)
cond_heatmap_canonica(-10, 10)
```


Empiricamente, percebemos que analisar valores de $d$ maiores que 80 não iam agregar na análise (a matriz de Vandermonde já fica muito mal condicionada a partir de 50, independente do intervalo) e valores de $m$ maiores que 100 também não alteram o padrão que vimos do condicionamento.

No mapa de calor, a escala foi truncada em $10^{20}$ para melhorar a visualização. Isto é, regiões com a cor máxima não indicam que o condicionamento é exatamente igual a $10^{20}$, mas sim que o condicionamento é da ordem de $10^{20}$ ou maior. Nos gráficos de linha (na letra e), sem o truncamento, podemos ver que em alguns casos o condicionamento pode atingir valores muito maiores, como $10^{50}$.

Os pontos brancos isolados correspondem a valores que o código não conseguiu representar bem na escala escolhida, geralmente por nan, inf ou só instabilidade numérica.

**ANÁLISE GRÁFICA:**

Note que a entrada (k,i) da matriz de Vandermonde vai ser $x_k^i$. Assim, se o valor de $d$ cresce, para os valores $x_i$ que têm módulo maior que 1, as potências $x_i^d$ explodem; e quando menor que 1 elas vão a zero, o que faz as colunas ficarem muito próximas ou muito distantes entre si, piorando o condicionamento.

Podemos ver isso explicitamente em cada um dos intervalos:

*1: 

Se $X \subset [-1, 1]$: Em módulo, todas as entradas são menores que 1, o que faz com que para um d $\approx$ 30, os vetores coluna associados a valores maiores de d são muito similares com o vetor nulo, explodindo o condicionamento de $V$;

*2:

Se $X \subset[-1, 0]$: Temos que todos os pontos $x_i$ são menores que zero. Isto é, além das potências irem a zero e as colunas a partir de um determinado valor de d ficarem todas similares (quase LD), as colunas ímpares e pares alternam de sinal, o que acaba também interferindo e piorando o condicionamento;

*3:

Se $X \subset[0, 1]$: Nesse caso, acontece o mesmo que no intervalo [-1, 0]: as colunas ficam muito similares entre si pelo módulo das entradas ser menor que 1, porém as colunas não vão alternar de sinal, explodindo o condicionamento;

*4: 

Se $X \subset[-10, 10]$: Nesse caso, estamos aumentando o tamanho dos pontos no intervalo, onde eles podem chegar a $10^d$ em módulo. Isso faz com que o maior valor singular exploda muito mais rapidamente, o que leva o condicionamento à aumentar de forma mais rápida com $d$ do que nos outros casos de intervalos que tomamos (no mapa de calor, ele é o único que atinge o valor do truncamento, em d $\approx$ 20 ).


Podemos concluir que o condicionamento das matrizes $L_X$ e $R_{x,d}$ na base canônica dependem fortemente do grau do polinômio e do intervalo selecionado, mas não dependem tanto do número de pontos m. A matriz de Vandermonde $V$ (consequentemente a $V^+$ também) é muito numericamente instável quando o valor de d cresce: perturbações pequenas no vetor b (erros de medição ou de arredondamento) podem gerar variações muito grandes nos coeficientes do polinômio ajustado, nos dando resultados não muito confiáveis.



**Comentários extras** (Falando sobre qualidade de ajuste do modelo e saindo um pouco dos condicionamentos):

Esses resultados nos levam a ver que $L_x$ e $R_{x,d}$ permanecem bem úteis independente do tamanho dos nossos dados de entrada. Entretanto, observando especificamente $R_{x,d}$ sob a perspectiva de $m \geq d+1$ observamos que para muitos valores de $m$ precisamos de polinômios cada vez mais complexos para obter uma aproximação satisfatória de $b$, ainda mais se consideramos que cada entrada $b_i$ de $b$ podem estar bem distantes uma da outra.

Isso implica que estaremos mais sujeitos a erros numéricos. Portanto, por mais que o valor de $m$ não influencie o condicionamento das tranformações quando cresce, para $R_{x,d}$, com $m$ grande e $d+1$ pequeno conseguimos resultados que talvez não nos sejam tão úteis.

### e)

Calcule as matrizes de $L_X$ e $R_{X,d}$ usando a base dos polinômios interpoladores de Lagrange.

Qual o condicionamento das matrizes $L_X$ e $R_{X,d}$ nestas bases?


Vamos analisar os dois casos das dimensões do domínio e contradomínio.

### 1) $m = d+1$.

Relembrando, temos que cada coluna da matriz de uma transformação linear é a imagem de um vetor da base do domínio. Assim, para calcular a matriz de $L_X$ na base dos polinômios interpoladores de Lagrange, basta calcular $L_X(p_i)$ para cada $i$.

Pela definição dos polinômios interpoladores de Lagrange, temos

$$
p_i(x_j)=\delta_{ij}.
$$

Logo,

$$
L_X(p_i)
=
\bigl(p_i(x_1),p_i(x_2),\dots,p_i(x_m)\bigr).
$$

No caso $m=d+1$, vamos ter

$$
L_X(p_i)
=
(0,\dots,0,1,0,\dots,0)
=
e_i,
$$

onde o $1$ aparece na $i$-ésima posição.

Assim, a matriz de $L_X$ na base dos polinômios interpoladores de Lagrange é

$$
\boxed{
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
1 & 0 & \cdots & 0\\
0 & 1 & \cdots & 0\\
\vdots & \vdots & \ddots & \vdots\\
0 & 0 & \cdots & 1
\end{pmatrix}
=
I_{d+1}.
}
$$

Relembrando, temos que o condicionamento de uma matriz inversível $A$ é dado por

$$
\kappa(A)=\|A\|\|A^{-1}\|.
$$

Como a gente viu, na base dos polinômios interpoladores de Lagrange, a matriz de $L_X$ é a própria identidade:

$$
[L_X]_{\mathcal{B}_{Lag}}=I.
$$

Assim,

$$
\kappa(I)=\|I\|\|I^{-1}\| =\|I\|^2.
$$

Agora vamos calcular $\|I\|$. Pela definição de norma induzida,
$$
\|I\|
=
\sup_{x\neq 0}
\frac{\|Ix\|}{\|x\|}
=
\sup_{x\neq 0}
\frac{\|x\|}{\|x\|}
=
1.
$$

Portanto,

$$
\kappa(I)
=
\|I\|^2
=
1^2
=
1.
$$

Assim,

$$
\kappa\left([L_X]_{\mathcal{B}_{Lag}}\right)=1.
$$

No caso da matriz de transformação de $R_{x,d}$, temos que encontrar o polinômio minimizador na base dos polinômios intepoladores de Lagrange. Vamos então aplicar a $R_{x,d}$ na base canônica do $\mathbb{R}^m$. Assim,

$$
R_{X,d}(e_i)
=
\operatorname{argmin}_{q\in P_d}
\|L_X(q)-e_i\|_2^2.
$$

Pela construção dos polinômios interpolares de Lagrange, vimos que:

$$
L_X(p_i)
=
(0,\dots,0,1,0,\dots,0)
=
e_i,
$$

Logo, temos que o polinômio minimizador $ R_{X,d}(e_i) $ será o próprio $p_i$. Mas, na base de Lagrange, o vetor de coordenadas de $p_i$ vai ser justamente $e_i$. Assim, como isso vale para todos os vetores da base canônica do $\mathbb{R}^m$, temos que:

$$
\boxed{
[R_{x,d}]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
| & | & & |\\
e_1 & e_2 & \cdots & e_m\\
| & | & & |
\end{pmatrix}
=
I_{d+1}
}
$$

Como vimos,  

$$
\kappa(I)=1.
$$

Logo, concluimos que quando $m = d+1$:

$$
\boxed{
\kappa\left([R_{x,d}]_{\mathcal{B}_{Lag}}\right)= \kappa[L_X]_{\mathcal{B}_{Lag}}
= 1.
}
$$


### 2) $ m>d+1 $:

Quando $m > d + 1$, temos mais pontos no conjunto $X$ = {${x_1, x_2, x_3, ... , x_m}$} do que polinômios na base de Lagrange. Não é possível definir uma base de $P_d$ com os polinômios interpoladores de Lagrange, já que dizer que:

$$
p_i(x_j)=\delta_{ij}.
$$

para $j = 1, 2, ..., m$ quando $m > d + 1$, estaremos impondo mais condições do que coeficientes disponíveis (temos até d+1 coeficientes). Assim, vamos tomar {${x_1, x_2, x_3, ... , x_{d+1}}$} pontos de $X$ distintos (teoricamente, podemos escolher qualquer subconjunto de d+1 pontos distintos de X para construir a base de Lagrange. Nos experimentos, comparamos duas escolhas: os primeiros d+1 pontos e d+1 pontos igualmente espaçados em X) e construir o conjunto na base de Lagrange {${p_1, p_2, p_3, ... , p_{d+1}}$} tal que:

$$
p_i(x_j)=\delta_{ij}.
$$

para $j = 1, 2, ..., d+1$. Assim, teremos uma base de Lagrange restrita à condição da dimensão do espaço $P_d$. Entretanto, como $L_x$ vai avaliar $p$ em todos os $m$ pontos de $X$, temos que:

$$
L_X(p_i)
=
\begin{pmatrix}
p_i(x_1)\\
\vdots\\
p_i(x_i)\\
\vdots\\
p_i(x_{d+1})\\
p_i(x_{d+2})\\
\vdots\\
p_i(x_m)
\end{pmatrix}.
$$

Então a coluna $i$ da representação matricial de $L_X$ é:

$$
L_X(p_i)
=
\begin{pmatrix}
0\\
\vdots\\
1\\
\vdots\\
0\\
p_i(x_{d+2})\\
\vdots\\
p_i(x_m)
\end{pmatrix}.
$$

Para que possamos calcular a matriz de $L_X$, precisamos saber quem são os polinômios na base de Lagrange aplicados em $x_{d+2},\dots,x_m$. Mas, temos uma fórmula fechada pra isso, que é:

$$

p_i(x)
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x-x_j}{x_i-x_j},
\qquad i=1,\dots,d+1.
$$

Portanto, a matriz de $L_X$ na base de Lagrange tem entradas

$$
L_{ki}
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x_k-x_j}{x_i-x_j}.
$$

Logo,

$$
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix},
$$

onde o bloco $*$ contém os valores

$$
p_i(x_k),
\qquad
k=d+2,\dots,m,
\quad
i=1,\dots,d+1.
$$


OBS:

Se esses nós forem os primeiros d+1 pontos de X, a matriz aparece explicitamente na forma 

$$
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix},
$$

mas se os nós forem escolhidos distribuídos ao longo do conjunto $X$, vamos ter a mesma estrutura, a menos de uma permutação das linhas da matriz. 

Vamos agora encontrar a representação matricial de $R_{x,d}$ quando $P_d$ está na base dos polinômios interpoladores de Lagrange. Para cada elemento da base canônica do $\mathbb{R}^m$, vamos encontrar $R_{x,d}(e_i)$. Nós vimos que a forma matricial de $R_{x,d}$ será a pseudoinversa de A:

$$
[R_{x,d}]_{\mathrm{Lag}}
=
A^+
$$

onde A é a representação matricial de $L_X$ se a matriz de $L_X$ tem posto completo nas colunas. Como a matriz de $L_X$ tem posto completo nas colunas, (um de seus blocos é a identidade $d+1$ x $d+1$) podemos afirmar que:

$$
\boxed{
[R_{x,d}]_{\mathrm{Lag}}
=
A^+
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix}
^+
}
$$

onde:

$$
A^+=(A^TA)^{-1}A^T.
$$

Porém, como calculamos os condicionamentos de $[R_{x,d}]_{\mathrm{Lag}}$ e de $[L_X]_{\mathrm{Lag}}$? Como vimos na letra d), podemos utilizar o fato de que:

$$
\kappa(A) = \frac{\sigma_{\max}}{\sigma_{\min}}
$$

onde $\sigma_{\max}$ e $\sigma_{\min}$ são o maior e menor valor singular, respectivamente.



Assim, vamos encontrar computacionalmente a fatoração SVD das matrizes $A$ = $[L_X]_{\mathrm{Lag}}$ e $A^+$ = $[R_{x,d}]_{\mathrm{Lag}}$ para diferentes valores de $m$ e $d$, como fizemos na letra anterior.  (na próxima célula):


```python

def matriz_L_lagrange(X, d, usar_primeiros_nos=False):
    X = np.asarray(X, dtype=float)
    if len(X) < d + 1:
        raise ValueError("Precisamos de m >= d+1.")
    m = len(X)
    r = d + 1

    if usar_primeiros_nos:
        indices = np.arange(r)
    else:
        indices = np.linspace(0, m - 1, r, dtype=int)
    
    nodes = X[indices]
    A = np.ones((m, r))
    for i in range(r):
        xi = nodes[i]
        for j in range(r):
            if j != i:
                xj = nodes[j]
                A[:, i] *= (X - xj) / (xi - xj)
    
    return A 

def calcular_grade_lagrange(
    x_menor,
    x_maior,
    m_min=2,
    m_max=150,
    passo_m=5,
    d_min=1,
    d_max=150,
    passo_d=5,
    usar_primeiros_nos=False,
    tol=1e-20,
):
    valores_m = list(range(m_min, m_max + 1, passo_m))
    valores_d = list(range(d_min, d_max + 1, passo_d))
    grade_conds = np.full((len(valores_m), len(valores_d)), np.nan)

    for idx_m, m in enumerate(valores_m):
        X = np.linspace(x_menor, x_maior, m)
        for idx_d, d in enumerate(valores_d):
            if m < d + 1:
                continue
            try:
                A = matriz_L_lagrange(X, d, usar_primeiros_nos=usar_primeiros_nos)
                if not np.all(np.isfinite(A)):
                    grade_conds[idx_m, idx_d] = np.inf
                else:
                    grade_conds[idx_m, idx_d] = cond_svd(A, tol=tol)
            except Exception:
                grade_conds[idx_m, idx_d] = np.inf

    return grade_conds, valores_m, valores_d


def plotar_heatmaps_lagrange(
    x_menor=-1, x_maior=1,
    m_min=2, m_max=150, passo_m=3,
    d_min=1, d_max=150, passo_d=3,
    vmin=1, vmax=1e20, tol=1e-15
):
    grade1, valores_m, valores_d = calcular_grade_lagrange(
        x_menor, x_maior, m_min, m_max, passo_m, d_min, d_max, passo_d,
        usar_primeiros_nos=False, tol=tol
    )
    grade2, _, _ = calcular_grade_lagrange(
        x_menor, x_maior, m_min, m_max, passo_m, d_min, d_max, passo_d,
        usar_primeiros_nos=True, tol=tol
    )

    fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)
    titulos = [
        "nós igualmente espaçados em $X$",
        "primeiros $d+1$ pontos de $X$"
    ]

    for ax, grade, titulo in zip(axes, [grade1, grade2], titulos):
        grade_plot = np.copy(grade)
        grade_plot[np.isinf(grade_plot)] = vmax
        grade_plot = np.ma.masked_invalid(grade_plot)
        
        D, M = np.meshgrid(valores_d, valores_m)
        
        im = ax.pcolormesh(
            D, M, grade_plot,
            cmap="viridis",
            norm=colors.LogNorm(vmin=vmin, vmax=vmax),
        )
        ax.set_title(titulo)
        ax.set_xlabel("grau do polinômio $d$")
        plt.colorbar(im, ax=ax, label="condicionamento")

    axes[0].set_ylabel("número de pontos $m$")
    fig.suptitle(
        f"comparação de escolha de nós na base de Lagrange\n"
        f"$X \\subset [{x_menor}, {x_maior}]$",
        fontsize=13
    )
    plt.tight_layout()
    plt.show()
```

```python
plotar_heatmaps_lagrange(
    x_menor=-1, x_maior=1,
    m_min=2, m_max=100, passo_m=1,
    d_min=1, d_max=100, passo_d=2,
    vmin=1, vmax=1e50)

plotar_heatmaps_lagrange(
    x_menor=0, x_maior=2,
    m_min=2, m_max=100, passo_m=1,
    d_min=1, d_max=100, passo_d=2,
    vmin=1, vmax=1e50)

plotar_heatmaps_lagrange(
    x_menor=-10, x_maior=10,
    m_min=2, m_max=100, passo_m=1,
    d_min=1, d_max=100, passo_d=2,
    vmin=1, vmax=1e50
)
```

**ANÁLISE GRÁFICA**

Primeiramente, o resultado que mais chama atenção é o que se refere a diagonal principal do gráfico, onde $m = d+1$, em que o condicionamento é 1 para todos os valores nessa reta, coincidindo com o que vimos anteriormente. 

Esse resultado para $m=d+1$ nos faz retomar a discussão do item 2(d), onde questionamos o quão favoráveis são as nossas transformações, principalmente da transformação $R_{x,d}$ onde eramos limitados a polinômios que não aproximavam $b$ de forma precisa se não quiséssemos lidar com altos valores de condicionamento. Nesse caso, a base de Lagrange é muito mais favorável numericamente, o que nos leva a pensar que o condicionamento da transformação depende bastante da base que tomamos para $P_d$.


Podemos perceber também que o condicionamento das matrizes depende fortemente da relação entre $d$, $m$ e da escolha dos nós que usamos pra construir a base. Isso já não acontece na base canônica, já que o crescimento de $d$ foi o fator dominante no comportamento do condicionamento.


Note que quando tomamos o conjunto $X$ com nós igualmente espaçados, no gráfico de calor, o crescimento do condicionamento parece ser mais afetado por $d$ do que por $m$. Isso acontece pois a matriz de $L_X$ tende a ficar mais equilibrada do que quando usamos só os primeiros d+1 pontos. Nesse caso, quando aumentamos m, estamos apenas adicioanndo mais linhas correspondentes a avaliar os mesmos polinômios em novos pontos, mas a base continua construída de forma distribuída no intervalo. 

Se tomamos os $d+1$ primeiros pontos de $X$, esses polinômios ficam "enviesados" pelos $d+1$ primeiro elementos do conjunto $X$, i.e., esses nós podem ficar concentrados em apenas uma parte do intervalo, principalmente quando $m >> d+1$. Nesse caso, estaremos construindo os polinômios para interpolar bem nessa região, mas se avaliarmos esses polinômios em outros pontos mais distantes, isso pode nos dar valores muito grandes pro bloco $*$.

Temos que na base de Lagrange, quando $m>d+1$, o condicionamento pode atingir valores mais altos quando comparado com a base canônica, se tomarmos a mesma variação de $d$ e $m$ nos dois gráficos. Isso acontece porque, mesmo que a parte associada aos nós da base forme uma identidade, quando avaliamos nos pontos extras, geramos o bloco $∗$ cujas entradas podem crescer bastante dependendo da escolha dos nós e de outros fatores que falamos mais cedo.

Ao variarmos os intervalos de $X$, temos que o condicionamento aparenta não mudar muito quando comparado com a base canônica. Esse fenômeno decorre do cálculo dos polinômios:

$$

p_i(x)
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x-x_j}{x_i-x_j},
\qquad i=1,\dots,d+1.
$$

Se todos os pontos forem escalados ou transladados, o numerador e denominador escalam ou transladam também e o resultado não muda. Assim, o condicionamento não depende tanto do intervalo escolhido, e sim mais da distribuição relativa deles. O mesmo não ocorre na base canônica, pois mudar $[−1,1]$ para $[−10,10]$ muda absurdamente as potências $x^d$.

Na próxima célula, vamos estudar a comparação entre o condicionamento das matrizes $L_X$ na base canônica e base de Lagrange fixando diferentes valores de $m$.

```python
def comparar_bases_m_fixo(X_intervalo=(-1, 1), valores_m=[50, 100, 200], d_max=150, tol=1e-15):
    
    x_menor, x_maior = X_intervalo
    d_vals = np.arange(1, d_max + 1)

    fig, axes = plt.subplots(1, len(valores_m), figsize=(6 * len(valores_m), 5), sharey=True)
    if len(valores_m) == 1:
        axes = [axes]

    for ax, m in zip(axes, valores_m):
        X = np.linspace(x_menor, x_maior, m)
        conds_vandermonde = []
        conds_lagrange = []

        for d in d_vals:
            if d + 1 > m:
                conds_vandermonde.append(np.nan)
                conds_lagrange.append(np.nan)
                continue

            # Base canônica — Vandermonde
            try:
                V = np.vander(X, d + 1, increasing=True)
                c = cond_svd(V, tol=tol)
                conds_vandermonde.append(c if np.isfinite(c) else np.nan)
            except Exception:
                conds_vandermonde.append(np.nan)

            # Base de Lagrange
            try:
                A = matriz_L_lagrange(X, d)
                c = cond_svd(A, tol=tol)
                conds_lagrange.append(c if np.isfinite(c) else np.nan)
            except Exception:
                conds_lagrange.append(np.nan)

        ax.semilogy(d_vals, conds_vandermonde, label="Base canônica (Vandermonde)", color="crimson")
        ax.semilogy(d_vals, conds_lagrange,    label="Base de Lagrange",            color="steelblue")
        ax.set_title(f"$m = {m}$")
        ax.set_xlabel("Grau do polinômio $d$")
        ax.legend()
        ax.grid(True, which="both", linestyle="--", alpha=0.5)

    axes[0].set_ylabel("Condicionamento (escala log)")
    fig.suptitle(
        f"Comparação de condicionamento: base canônica vs Lagrange\n"
        f"$X \\subset [{x_menor}, {x_maior}]$",
        fontsize=13
    )
    plt.tight_layout()
    plt.show()
```

```python
comparar_bases_m_fixo(X_intervalo=(-1, 1))
comparar_bases_m_fixo(X_intervalo=(0, 2))
comparar_bases_m_fixo(X_intervalo=(-10, 10))
```

Vale ressaltar que a matriz de Lagrange está sendo calculada com os nós igualmente espaçados no conjunto $X$, de forma que deixemos a matriz $L_X$ mais equilibrada.

### f)

Compare a complexidade computacional de calcular $L_X$ e $R_{X,d}$ usando as bases canônica e de Lagrange com o condicionamento destas matrizes.

---

Vamos analisar a complexidade computacional de calcular $L_X$ e $R_{x,d}$ em cada uma das bases:

### 1) Base canônica:

Analisando a matriz de $L_X$:

Temos que a matriz de $L_X$ é $V$, onde a matriz V é a matriz de Vandermonde associada aos pontos de $X$. Note que na $i$-ésima linha de V, estamos calculando potências de $x_i$ de grau até $d$. Assim, em cada linha, a complexidade é de até $O(d+1)$. Como isso vale para todas as $m$ linhas da matriz V, temos que a complexidade computacional de calcular $L_X$ é:

$$
\boxed{
L_X \rightarrow
O(m(d+1))
}
$$

Analisando a matriz de $R_{x,d}$:

Na matriz de $R_{x,d}$, temos que calcular a pseudoinversa da matriz de Vandermonde $V$. Podemos prosseguir de três formas diferentes:

**1. Por meio da fatoração reduzida SVD de $V$:**

A complexidade computacional de calcular a fatoração SVD reduzida no NumPy (fazendo a análise assintótica apenas) de uma matriz $m\times(d+1)$ quando $m$ $\geq$ $d+1$, é de 

$$
O(m(d+1)^2)
$$ 

Sabendo a fatoração reduzida de V:

$$
V
=
\hat{U}\hat{\Sigma} \hat{W}^T
$$

A pseudoinversa de V será:

$$
V^+
=
\hat{W}\hat{\Sigma}^{-1} \hat{U}^T
$$

Como encontramos a SVD reduzida de $V$, temos que:

$$
\hat{\Sigma}
=
\begin{pmatrix}
\operatorname{diag}(\sigma_0,\sigma_1,\dots,\sigma_{d})\\
\end{pmatrix}.
$$

Então, a matriz inversa será:

$$
\hat{\Sigma}^{-1}

=
\operatorname{diag}
\left(
\frac{1}{\sigma_{0}},
\frac{1}{\sigma_{1}},
\dots,
\frac{1}{\sigma_d}
\right).
$$


Calcular essa pseudoinversa tem a complexidade computacional de inverter cada entrada dos elementos da diagonal (custo $\approx$ $O(d+1)$), que tem ordem menor que a complexidade de calcular a SVD reduzida de $V$. Assim, vimos que a etapa mais custosa vai ser de calcular a fatoração SVD reduzida de $V$, com custo 

$$
O(m(d+1)^2)
$$

Se quisermos explicitar a matriz $V^+$ e calcular o produto das matrizes, a complexidade computacional fica na mesma ordem de $O(m(d+1)^2)$, ou até menor. 

Por fim, a complexidade computacional de calcular a $R_{x,d}$ utilizando a fatoração SVD reduzida de $V$ é:

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$


**2. Por meio da fatoração QR reduzida de $V$:**

Sabemos que, se a matriz $V$ tem fatoração QR reduzida:

$$
V=\hat Q\hat R,
$$

Como visto em aula, a pseudoinversa de $V$ será (se a matriz $V$ tem posto completo, nosso caso):

$$

V^+
=
\hat R^{-1}\hat Q^T.
$$

Agora vamos analisar a complexidade computacional. A fatoração QR reduzida de uma matriz $V$ $m\times(d+1)$, quando $m \geq d+1$ tem complexidade computacional da ordem de

$$
O(m(d+1)^2).
$$

Se quisermos formar explicitamente a matriz

$$
V^+=\hat R^{-1}\hat Q^T,
$$

não precisamos calcular $\hat R^{-1}$ diretamente. Podemos resolver o sistema triangular

$$
\hat R X=\hat Q^T.
$$

Como $\hat R$ tem dimensão $(d+1)\times(d+1)$ e $\hat Q^T$ tem dimensão $(d+1)\times m$, resolver esse sistema triangular com $m$ lados direitos tem complexidade da ordem de

$$
O(m(d+1)^2).
$$

Assim, a etapa de formar explicitamente $V^+$ também fica na ordem de

$$
O(m(d+1)^2).
$$

Logo, a complexidade computacional para calcular $R_{X,d}$ usando a fatoração QR reduzida de $V$ fica na ordem de

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$

Assintoticamente, QR reduzida e SVD reduzida ficam na mesma ordem $O(m(d+1)^2)$ quando $m \geq d+1$. Porém, QR costuma ser computacionalmente mais barata (matrizes ortogonais), enquanto SVD é mais informativa e mais robusta para analisar posto, valores singulares e condicionamento.


**3. Calcular o produto $V^+ = (V^TV)^{-1}V^T$;**

A complexidade computacional de calcular $V^TV$ é (produto matricial)

$$
O(m(d+1)^2)
$$

A partir desse produto, temos dois caminhos:


i) Resolver o sistema:

$$
(V^TV)X
=
V^T
$$

ii) Resolver a equação:

$$
x
=
(V^TV)^{-1}V^T
$$

Ao resolver o sistema (i) por fatoração (LU), temos que a complexidade é de 

$$
O((d+1)^3)
$$

e a substituição vai ter complexidade:

$$
O(m(d+1)^2)
$$

Assim, a complexidade final será:

$$
O(m(d+1)^2 + m(d+1) + (d+1)^3 + m(d+1)^2) \approx O(m(d+1)^2 + (d+1)^3)
$$

Ou seja (já que $m \geq d+1$),

$$
O(m(d+1)^2)
$$


Resolvendo a equação em (ii), obteríamos que calcular a inversa ($V^TV$ tem dimensões $(d+1)\times{(d+1)}$) tem complexidade computacional:

$$
O((d+1)^3)
$$

Depois, fazer o produto de $(V^TV)^{-1}$ com ${V^T}$ tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Assim, a complexidade computacional é a mesma:

$$
O(m(d+1)^2 + (d+1)^3 + m(d+1)^2) =  O(m(d+1)^2 + (d+1)^3)
$$

isto é,

$$
O(m(d+1)^2)
$$

Assim, de ambos os métodos, podemos concluir que a complexidade computacional de calcular a matriz de $R_{x,d}$ por meio da fórmula fechada da pseudoinversa é:

$$
R_{x,d} \rightarrow
O(m(d+1)^2)
$$

Assim, assintoticamente falando, temos que a complexidade computacional de calcular a matriz $R_{x,d}$ é de 

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$

OBS: Apesar de assintoticamente todos os métodos serem iguais, temos que cada um acompanha uma constante diferente e cada um deles garantem propriedades diferentes. Por exemplo, a SVD é a fatoração mais informativa, porém a fatoração QR chega a ser a mais barata dentre todas. Já o método do produto nos faz calcular a matriz quadrada $V^TV$, que o condicionamento é aprox. o quadrado do condicionamento de $V$, o que é péssimo se a matriz $V$ já está mal condicionada.

### 2) Base dos Polinômios Interpoladores de Lagrange.

Na base dos polinômios interpoladores de Lagrange, precisamos analisar os dois casos das dimensões do domínio e contradomínio:

**i) $m$ = $d+1$**

Na base de Lagrange, temos que a matriz de $L_X$ e de $R_{x,d}$ são do formato:

$$
[R_{x,d}]_{\mathcal{B}_{Lag}}= [L_X]_{\mathcal{B}_{Lag}}
= I_{d+1}
$$

Isso implica que a complexidade computacional de montar explicitamente as matrizes de $L_X$ e $R_{x,d}$ é:

$$
\boxed{
R_{x,d}, L_X \rightarrow
O((d+1)^2)
}
$$

Se tratarmos a identidade de forma implícita, não precisamos montar nem multiplicar a matriz. Nesse caso, o custo computacional vai ser de $O(1)$. 

**ii) $m > d+1$:**

Vimos que a matriz de $L_X$ quando $m > d+1$:

$$
[L_X]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix},
$$

e a matriz de $R_{x,d}$ será:

$$
[R_{x,d}]_{\mathcal{B}_{Lag}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix} ^+
$$

onde o bloco $*$ contém os valores

$$
p_i(x_k),
\qquad
k=d+2,\dots,m,
\quad
i=1,\dots,d+1.
$$


A complexidade computacional de calcular cada entrada é a complexidade de calcular o produtório:

$$
p_i(x_k)
=
\prod_{\substack{j=1\\ j\neq i}}^{d+1}
\frac{x_k-x_j}{x_i-x_j},
\qquad i=1,\dots,d+1.
$$

que tem complexidade:

$$
O(d+1)
$$

Temos que o bloco $*$ tem $m-(d+1)$ linhas e $d+1$ colunas, o que nos diz que a complexidade de calcular o bloco $*$ terá complexidade computacional:

$$
O(d(m-d-1)(d+1))
$$

Assintoticamente, quando $m > d+1$:

$$
O(d(m-d-1)(d+1)) = O(m(d+1)^2)
$$

Calcular o bloco identidade $I^{d+1}$ tem complexidade $O((d+1)^2)$. Assim, a complexidade computacional total de calcular a matriz $L_X$ é:

$$
\boxed{
L_X \rightarrow
O(m(d+1)^2)
}
$$

Vamos analisar a pseudoinversa da matriz de $L_X$, isto é, a matriz de $R_{x,d}$. 

Para calcularmos a matriz de $R_{x,d}$, precisamos primeiro montar a matriz de $L_X$ e depois calcular a sua pseudoinversa. Vimos que montar a matriz de $L_X$ tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Provamos anteriormente que a complexidade computacional de calcular $A^+$ (quando A tem posto completo) pela fatoração QR, SVD e equações normais tem complexidade computacional:

$$
O(m(d+1)^2)
$$

Assim, como a matriz de $L_X$ tem posto completo, temos que:

$$
\boxed{
R_{x,d} \rightarrow
O(m(d+1)^2)
}
$$


Vamos relacionar os condicionamentos dessas matrizes:


| Base | Matriz de $L_X$ | Custo para calcular $L_X$ | Matriz de $R_{X,d}$ | Custo para calcular $R_{X,d}$ | Condicionamento |
|---|---|---:|---|---:|---|
| Canônica | Vandermonde $V$ | $O(m(d+1))$ | $V^+$ | $O(m(d+1)^2)$ | É muito ruim quando $d$ cresce |
| Lagrange, $m=d+1$ | $I_{d+1}$ | Baixo, ou $O((d+1)^2)$ se construirmos a $I_{d+1}$ | $I_{d+1}$ | Baixo, ou $O((d+1)^2)$ se construirmos a $I_{d+1}$  | $\kappa(I)=1$ |
| Lagrange, $m>d+1$ | $\begin{pmatrix} I_{d+1}\\ * \end{pmatrix}$ | $O(m(d+1)^2)$ usando a fórmula direta | $A^+$, onde $A=[L_X]_{\mathcal B_{\mathrm{Lag}}}$ | $O(m(d+1)^2)$ | Depende dos nós escolhidos; pode crescer quando $d$ aumenta |

A base canônica é barata para montar $L_X$, porque é só calcular as potências

$$
1,x_i,x_i^2,\dots,x_i^d.
$$

Assim, a matriz de $L_X$ na base canônica é a matriz de Vandermonde $V$, e seu custo de montagem é da ordem de

$$
O(m(d+1)).
$$

Porém, a matriz de Vandermonde costuma ficar muito mal condicionada quando o grau $d$ aumenta. Portanto, o baixo custo de montagem não significa boa estabilidade numérica.

Já a base de Lagrange é extremamente favorável no caso em que $m=d+1$. Nesse caso, 

$$
[L_X]_{\mathcal B_{\mathrm{Lag}}}=[R_{X,d}]_{\mathcal B_{\mathrm{Lag}}}=I_{d+1}.
$$

Logo, o condicionamento é ideal:

$$
\kappa(I)=1.
$$

Porém, quando $m>d+1$, aparece o bloco adicional

$$
[L_X]_{\mathcal B_{\mathrm{Lag}}}
=
\begin{pmatrix}
I_{d+1}\\
*
\end{pmatrix}.
$$

Nesse caso, precisamos avaliar os polinômios de Lagrange nos pontos extras. Usando a fórmula direta,

$$
p_i(x_k)
=
\prod_{\substack{j=1\\j\neq i}}^{d+1}
\frac{x_k-x_j}{x_i-x_j},
$$

cada entrada do bloco $*$ custa $O(d)$ para ser calculada. Como há aproximadamente $m(d+1)$ entradas, o custo de montar essa matriz fica da ordem de

$$
O(m(d+1)^2).
$$

Depois, para calcular $R_{X,d}$ nessa base, precisamos calcular a pseudoinversa de

$$
A=[L_X]_{\mathcal B_{\mathrm{Lag}}},
$$

isto é,

$$
[R_{X,d}]_{\mathcal B_{\mathrm{Lag}}}=A^+.
$$

Usando SVD ou QR reduzida, esse custo fica da ordem de

$$
O(m(d+1)^2).
$$

Como a complexidade computacional vai medir o custo para calcular as matrizes e o condicionamento mede a sensibilidade numérica dessas matrizes, temos que uma base pode ser barata de montar, mas a mesma pode ser numericamente ruim, como a base canônica. Por outro lado, uma base pode ser muito bem condicionada em certos casos, como a base de Lagrange quando $m=d+1$, mas pode ser mais custosa de montar quando $m>d+1$. 

Vale notar que na base canônica vamos ter uma assimetria, já que montar a matriz $L_X$ é barato ($O(m(d+1))$), mas calcular a $R_{x,d}$ custa $O(m(d+1)^2)$ e mesmo pagando esse custo maior, o condicionamento das duas continua péssimo conforme d cresce. Já na base de Lagrange com m=d+1, as duas matrizes são $I_{d+1}$, com o mesmo custo e com o mesmo condicionamento = 1.
