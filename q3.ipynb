## Questão 3. Interpolação e Extrapolação

Considere

$$
X=\operatorname{linspace}(-1,1,m),
$$

ou seja, $m$ pontos igualmente espaçados em $[-1,1]$, e, analogamente,

$$
Y=\operatorname{linspace}(-1,1,n).
$$

Continuamos supondo $m\geq d+1$.


### a)

Qual o posto da aplicação composta

$$
L_Y\circ R_{X,d}?
$$

Pela questão 2(b) temos: $L_Y = V_y$, onde a matriz $V_y$ é uma Vandermonde aplicada nos pontos de $Y$ e $R_{X,d} = V_x^{+}$, a pseudoinversa da matriz de Vandermonde associada a $L_X$.


Portanto, a composição das duas transformações lineares vai ser:


$$
L_Y\circ R_{X,d} = V_YV_x^{+}
$$


A partir disso, temos que o posto do produto de duas matrizes nunca excede o posto de cada uma delas individualmente, i.e.:

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\operatorname{rank}(V_Y),\operatorname{rank}(V_X^+)\}.
$$

Mas o posto da pseudo inversa de $V_x$ é igual ao posto da própria $V_x$. Já que $ m \geq d+1 $, o posto de $V_x$ será

$$
\operatorname{rank}(V_X)=\min\{m,d+1\}=d+1.
$$

Logo:

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\operatorname{rank}(V_Y),d+1\}.
$$


E como 

$$
\operatorname{rank}(V_Y)=\min\{n,d+1\},
$$

temos

$$
\operatorname{rank}(V_YV_X^+)
\leq
\min\{\min\{n,d+1\},d+1\}.
$$

$$
\Rightarrow
rank(V_y V_x^+) \leq min\{n, d + 1\}

$$

Vamos além! Como $m \geq d+1$, a transformação 

$$
R_{X,d}:\mathbb R^m\to P_d
$$


é sobrejetiva, i.e.:

$$
\operatorname{Im}(R_{X,d})=P_d.
$$

Logo,

$$
\operatorname{rank}(L_Y\circ R_{X,d})
=
\dim(\operatorname{Im}(L_Y\circ R_{X,d}))
=
\dim(\operatorname{Im}(L_Y))
=
\operatorname{rank}(L_Y).
$$

Como

$$
\operatorname{rank}(L_Y)=\min\{n,d+1\},
$$

concluímos que

$$
\boxed{
\operatorname{rank}(L_Y\circ R_{X,d})
=
\min\{n,d+1\}.
}
$$

### b)

Explique porque a composta acima não depende da base escolhida para $P_d$.

Fixe uma base b de $P_d$. Queremos provar que para qualquer outra base $C$ de $P_d$, a composta independe da base escolhida, isto é:

$$
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
=
[L_Y]_{\mathcal{C}}[R_{x,d}]_{\mathcal{C}}
$$

Se $p$ é um polinômio em $P_d$, podemos denotar o mesmo polinômios em duas bases:

$$
[p]_{\mathcal{B}} , [p]_{\mathcal{C}}
$$

Logo, vai existir uma matriz $S$ de mudança de base tal que:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Vamos encontrar a matriz de $L_Y$ na base C. Sabemos que:

$$
L_y(p) = [L_Y]_{\mathcal{B}}[p]_{\mathcal{B}}
$$

Entretanto, podemos reescrever isso como:

$$
L_y(p) = [L_Y]_{\mathcal{B}}S[p]_{\mathcal{C}}
$$

Como a transformação $L_Y$ aplicada no mesmo polinômio p mas na base C é:

$$
L_y(p) = [L_Y]_{\mathcal{C}}[p]_{\mathcal{C}}
$$

Podemos concluir que:

$$
[L_Y]_{\mathcal{C}} = [L_Y]_{\mathcal{B}}S
$$

Vamos olhar agora para $R_{x,d}$ na base C. Sabemos que para cada $b$, o vetor $[R_{x,d}]b_{\mathcal{B}}$ denota os coeficientes do polinômio minimizador associado a $b$ na base $B$. Entretanto, o vetor $[R_{x,d}]b_{\mathcal{B}}$ irá retornar os coeficientes do mesmo polinômio minimizador, mas na base $C$. Como as coordenadas do mesmo polinômio em bases diferentes se relacionam por:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Então, temos que:

$$
[R_{x,d}b]_{\mathcal{B}} = S[R_{x,d}b]_{\mathcal{C}}
$$

Invertendo a matriz $S$, concluímos que:

$$
[R_{x,d}b]_{\mathcal{C}} = S^{-1}[R_{x,d}b]_{\mathcal{C}}
$$

Agora que temos as matrizes de $L_Y$ e $R_{x,d}$ na base $C$, vamos compôr essas transformações:

$$
L_Y\circ R_{X,d} 
=
[L_Y]_{\mathcal{C}} [R_{x,d}b]_{\mathcal{C}}
=
([L_Y]_{\mathcal{B}}S) (S^{-1}[R_{x,d}b]_{\mathcal{C}})
=
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
$$

Assim, conseguimos provar que a matriz da composta independe da base.


### c)

Em que sentido esta composta é uma aplicação de interpolação?

Sabemos que a interpolação polinomial consiste em construir um polinômio que passa exatamente por um conjunto de pontos dados. Tomando um par de vetores $x,\ b \in \R^m, \ x_i \in X$ tal que o par ordenado $(x_i,\ b_i)$ representa um ponto em $\R^2$, se quisermos interpolar esses dados podemos construir uma curva que passe por esses pontos fazendo da seguinte forma:


Mandamos o vetor de dados para $R_{x,d}$ e recebemos um polinômio que melhor aproxima eles.

Temos que esse polinômio aplicado ao avaliador $L_X$ é o mais próximo possível de $b$ na norma euclidiana. Vamos tomar outro conjunto de pontos e aplicar $L_Y$ ao polinômio ajustado. Isso nos dá o vetor:

$$
(p(y_1),\dots,p(y_n)).
$$

Isto é, realizamos $n$ novas amostragens da curva $p(x)$ através de $L_y$, nos dando o gráfico da curva nos pontos $(y_i, p(y_i))$.

A composta $L_Y \circ R_{x,d}$ pega os dados observados nos pontos $X$, ajusta um polinômio de grau no máximo $d$ por mínimos quadrados e depois avalia esse polinômio em novos pontos $Y$.

Quando os dados pertencem à imagem de $L_x$, temos que existe um polinômio que ajusta precisamente os pontos $x_i$, o que se encaixa em uma interpolação. Porém, quando $m \geq d+1$, existem mais pontos do que coeficientes, o que torna impossível de existir uma interpolação para o conjunto de dados $X$.

### d)

Escolha alguns valores de $m$, $n$ e $d$ e calcule os valores singulares da composta, usando como intermediárias tanto a base canônica quanto a base dos polinômios interpoladores de Lagrange.

Qual a diferença entre as matrizes calculadas?

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors


def cond_efetivo(A, tol=1e-12):
    """
    Condicionamento usando apenas valores singulares numericamente não nulos.
    Útil quando a matriz é retangular ou posto-deficiente.
    """
    sigma = np.linalg.svd(A, compute_uv=False)

    sigma_nonzero = sigma[sigma > tol * sigma[0]]

    if len(sigma_nonzero) == 0:
        return np.inf

    return sigma_nonzero[0] / sigma_nonzero[-1]


def lagrange_eval(t, nodes, i):
    produto = 1.0
    zi = nodes[i]

    for j, zj in enumerate(nodes):
        if j != i:
            produto *= (t - zj) / (zi - zj)

    return produto


def matriz_avaliacao_lagrange(pontos, nodes):
    pontos = np.asarray(pontos, dtype=float)
    nodes = np.asarray(nodes, dtype=float)

    n = len(pontos)
    r = len(nodes)

    A = np.zeros((n, r))

    for k in range(n):
        for i in range(r):
            A[k, i] = lagrange_eval(pontos[k], nodes, i)

    return A


def composta_canonica(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    V_X = np.vander(X, N=d+1, increasing=True)
    V_Y = np.vander(Y, N=d+1, increasing=True)

    C = V_Y @ np.linalg.pinv(V_X, rcond=rcond)

    return C


def composta_lagrange(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    # Mesma base de Lagrange para X e Y
    nodes = np.linspace(intervalo_X[0], intervalo_X[1], d+1)

    A_X = matriz_avaliacao_lagrange(X, nodes)
    A_Y = matriz_avaliacao_lagrange(Y, nodes)

    C = A_Y @ np.linalg.pinv(A_X, rcond=rcond)

    return C


m = 100
n = 100

valores_d = list(range(1, m))

conds_can = []
conds_lag = []
diffs = []

for d in valores_d:
    C_can = composta_canonica(m, n, d)
    C_lag = composta_lagrange(m, n, d)

    conds_can.append(cond_efetivo(C_can))
    conds_lag.append(cond_efetivo(C_lag))
    diffs.append(np.max(np.abs(C_can - C_lag)))


plt.figure(figsize=(9, 6))

plt.plot(valores_d, conds_can, label="Base canônica", marker="o", markersize=3)
plt.plot(valores_d, conds_lag, label="Base de Lagrange", linestyle="--")

plt.yscale("log")

plt.xlabel("Grau do polinômio d")
plt.ylabel("Condicionamento efetivo")
plt.title(f"Condicionamento efetivo de $L_Y \\circ R_{{X,d}}$ para m={m}, n={n}")

plt.legend()
plt.grid(True, which="both", alpha=0.3)

plt.show()
```

```python
def heatmap_composta_m_fixo(
    m=70,
    n_min=10,
    n_max=150,
    passo_n=5,
    d_min=1,
    d_max=None,
    passo_d=2,
    base="canonica",
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    rcond=1e-12,
    vmax=1e16
):
    if d_max is None:
        d_max = m - 1

    valores_n = list(range(n_min, n_max + 1, passo_n))
    valores_d = list(range(d_min, d_max + 1, passo_d))

    grade = np.full((len(valores_n), len(valores_d)), np.nan)

    for idx_n, n in enumerate(valores_n):
        for idx_d, d in enumerate(valores_d):

            if m < d + 1:
                continue

            try:
                if base == "canonica":
                    C = composta_canonica(
                        m, n, d,
                        intervalo_X=intervalo_X,
                        intervalo_Y=intervalo_Y,
                        rcond=rcond
                    )
                elif base == "lagrange":
                    C = composta_lagrange(
                        m, n, d,
                        intervalo_X=intervalo_X,
                        intervalo_Y=intervalo_Y,
                        rcond=rcond
                    )
                else:
                    raise ValueError("base deve ser 'canonica' ou 'lagrange'.")

                grade[idx_n, idx_d] = cond_efetivo(C)

            except Exception:
                grade[idx_n, idx_d] = np.inf

    grade_plot = np.copy(grade)
    grade_plot[np.isinf(grade_plot)] = vmax

    plt.figure(figsize=(10, 8))

    im = plt.imshow(
        grade_plot,
        extent=[valores_d[0], valores_d[-1], valores_n[0], valores_n[-1]],
        origin="lower",
        cmap="viridis",
        norm=colors.LogNorm(vmin=1, vmax=vmax),
        aspect="auto",
        interpolation="nearest"
    )

    plt.colorbar(im, label="Condicionamento efetivo")
    plt.xlabel("Grau do polinômio d")
    plt.ylabel("Número de pontos n em Y")
    plt.title(f"Condicionamento efetivo de $L_Y \\circ R_{{X,d}}$ com m={m} — base {base}")

    plt.show()

    return valores_n, valores_d, grade
```

```python
def heatmap_diferenca_compostas(
    m=70,
    n_min=10,
    n_max=150,
    passo_n=5,
    d_min=1,
    d_max=None,
    passo_d=2,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    rcond=1e-12,
    vmin=1e-16,
    vmax=1e2
):
    if d_max is None:
        d_max = m - 1

    valores_n = list(range(n_min, n_max + 1, passo_n))
    valores_d = list(range(d_min, d_max + 1, passo_d))

    grade_diferencas = np.full((len(valores_n), len(valores_d)), np.nan)

    for idx_n, n in enumerate(valores_n):
        for idx_d, d in enumerate(valores_d):

            # Hipótese do problema: m >= d+1
            if m < d + 1:
                continue

            try:
                C_can = composta_canonica(
                    m=m,
                    n=n,
                    d=d,
                    intervalo_X=intervalo_X,
                    intervalo_Y=intervalo_Y,
                    rcond=rcond
                )

                C_lag = composta_lagrange(
                    m=m,
                    n=n,
                    d=d,
                    intervalo_X=intervalo_X,
                    intervalo_Y=intervalo_Y,
                    rcond=rcond
                )

                diferenca = np.max(np.abs(C_can - C_lag))

                grade_diferencas[idx_n, idx_d] = diferenca

            except Exception:
                grade_diferencas[idx_n, idx_d] = np.nan

    grade_plot = np.copy(grade_diferencas)

    # LogNorm não aceita zero, então substituímos zeros por vmin
    grade_plot[grade_plot == 0] = vmin

    plt.figure(figsize=(10, 8))

    cmap = plt.cm.viridis.copy()
    cmap.set_bad(color="lightgray")

    im = plt.imshow(
        grade_plot,
        extent=[valores_d[0], valores_d[-1], valores_n[0], valores_n[-1]],
        origin="lower",
        cmap=cmap,
        norm=colors.LogNorm(vmin=vmin, vmax=vmax),
        aspect="auto",
        interpolation="nearest"
    )

    plt.colorbar(im, label=r"$\max |C_{\mathrm{can}}-C_{\mathrm{Lag}}|$")
    plt.xlabel("Grau do polinômio d")
    plt.ylabel("Número de pontos n em Y")
    plt.title(
        f"Diferença entre as compostas nas bases canônica e de Lagrange\n"
        f"m={m}, X⊂[{intervalo_X[0]}, {intervalo_X[1]}], Y⊂[{intervalo_Y[0]}, {intervalo_Y[1]}]"
    )

    plt.show()

    return valores_n, valores_d, grade_diferencas
```

### e)

Qual o condicionamento da composta

$$
L_Y\circ R_{X,d}?
$$

Você pode fazer $n\to\infty$ sem prejudicar o condicionamento?

E $m\to\infty$ e $d\to\infty$?

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as colors

valores_m = list(range(2, 150))
valores_d = list(range(1, 80))

grade_conds = np.full((len(valores_m), len(valores_d)), np.nan)

def cond_svd(A, tol=1e-15):
    sigma = np.linalg.svd(A, compute_uv=False)

    if sigma[-1] < tol:
        return np.inf

    return sigma[0] / sigma[-1]

def cond_heatmap_canonica(x_menor, x_maior):
    for idx_m, m in enumerate(valores_m):

        X = np.linspace(x_menor, x_maior, m)

        for idx_d, d in enumerate(valores_d):

            # A questão supõe m >= d + 1
            if m < d + 1:
                continue

            # P_d tem dimensão d+1: colunas 1, x, x^2, ..., x^d
            V = np.vander(X, N=d+1, increasing=True)

            try:
                grade_conds[idx_m, idx_d] = cond_svd(V)
            except np.linalg.LinAlgError:
                grade_conds[idx_m, idx_d] = np.inf


    # Para o mapa de calor ficar legível, usamos escala logarítmica.
    # Também cortamos valores infinitos para um teto visual.
    grade_plot = np.copy(grade_conds)
    grade_plot[np.isinf(grade_plot)] = 1e16

    plt.figure(figsize=(10, 8))

    im = plt.imshow(
        grade_plot,
        extent=[valores_d[0], valores_d[-1], valores_m[0], valores_m[-1]],
        origin="lower",
        cmap="viridis",
        norm=colors.LogNorm(vmin=1, vmax=1e16),
        aspect="auto",
    )

    plt.colorbar(im, label="$cond(L_X)$")
    plt.xlabel("grau do polinômio d")
    plt.ylabel("número de pontos m")
    plt.title(f"cond($L_X$) = cond($R_xd$) na base canônica em linspace({x_menor}, {x_maior}, m)")

    plt.show()
```
