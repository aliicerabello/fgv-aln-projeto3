## Questão 3. Interpolação e Extrapolação

Considere

$$
X=\operatorname{linspace}(-1,1,m),
$$

ou seja, $m$ pontos igualmente espaçados em $[-1,1]$, e, analogamente,

$$
Y=\operatorname{linspace}(-1,1,n).
$$

Continuamos supondo $m\geq d+1$.


### a)

Qual o posto da aplicação composta

$$
L_Y\circ R_{X,d}?
$$

Como $m \geq d+1$, a transformação 

$$
R_{X,d}:\mathbb R^m\to P_d
$$


é sobrejetiva (da questão 2a), para qualquer $p \in P_d$, basta tomar $b = L_X(p) \Rightarrow R_{x,d}(b) = p$ ), i.e.:

$$
\operatorname{Im}(R_{X,d})=P_d.
$$

Logo,

$$
\operatorname{rank}(L_Y\circ R_{X,d})
=
\dim(\operatorname{Im}(L_Y\circ R_{X,d}))
=
\dim(\operatorname{Im}(L_Y))
=
\operatorname{rank}(L_Y).
$$

Como

$$
\operatorname{rank}(L_Y)=\min\{n,d+1\},
$$

concluímos que

$$
\boxed{
\operatorname{rank}(L_Y\circ R_{X,d})
=
\min\{n,d+1\}.
}
$$

$$



### b)

Explique porque a composta acima não depende da base escolhida para $P_d$.

Fixe uma base b de $P_d$. Queremos provar que para qualquer outra base $C$ de $P_d$, a composta independe da base escolhida, isto é:

$$
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
=
[L_Y]_{\mathcal{C}}[R_{x,d}]_{\mathcal{C}}
$$

Se $p$ é um polinômio em $P_d$, podemos denotar o mesmo polinômios em duas bases:

$$
[p]_{\mathcal{B}} , [p]_{\mathcal{C}}
$$

Logo, vai existir uma matriz $S$ de mudança de base tal que:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Vamos encontrar a matriz de $L_Y$ na base C. Sabemos que:

$$
L_y(p) = [L_Y]_{\mathcal{B}}[p]_{\mathcal{B}}
$$

Entretanto, podemos reescrever isso como:

$$
L_y(p) = [L_Y]_{\mathcal{B}}S[p]_{\mathcal{C}}
$$

Como a transformação $L_Y$ aplicada no mesmo polinômio p mas na base C é:

$$
L_y(p) = [L_Y]_{\mathcal{C}}[p]_{\mathcal{C}}
$$

Podemos concluir que:

$$
[L_Y]_{\mathcal{C}} = [L_Y]_{\mathcal{B}}S
$$

Vamos olhar agora para $R_{x,d}$ na base C. Sabemos que para cada $b$, o vetor $[R_{x,d}]b_{\mathcal{B}}$ denota os coeficientes do polinômio minimizador associado a $b$ na base $B$. Entretanto, o vetor $[R_{x,d}]b_{\mathcal{B}}$ irá retornar os coeficientes do mesmo polinômio minimizador, mas na base $C$. Como as coordenadas do mesmo polinômio em bases diferentes se relacionam por:

$$
[p]_{\mathcal{B}} = S[p]_{\mathcal{C}}
$$

Então, temos que:

$$
[R_{x,d}b]_{\mathcal{B}} = S[R_{x,d}b]_{\mathcal{C}}
$$

Invertendo a matriz $S$, concluímos que:

$$
[R_{x,d}b]_{\mathcal{C}} = S^{-1}[R_{x,d}b]_{\mathcal{B}}
$$

Agora que temos as matrizes de $L_Y$ e $R_{x,d}$ na base $C$, vamos compôr essas transformações:

$$
L_Y\circ R_{X,d} 
=
[L_Y]_{\mathcal{C}} [R_{x,d}b]_{\mathcal{C}}
=
([L_Y]_{\mathcal{B}}S) (S^{-1}[R_{x,d}b]_{\mathcal{C}})
=
[L_Y]_{\mathcal{B}}[R_{x,d}]_{\mathcal{B}}
$$

Assim, conseguimos provar que a matriz da composta independe da base.


### c)

Em que sentido esta composta é uma aplicação de interpolação?

Sabemos que a interpolação polinomial consiste em construir um polinômio que passa exatamente por um conjunto de pontos dados. Tomando um par de vetores $x,\ b \in \R^m, \ x_i \in X$ tal que o par ordenado $(x_i,\ b_i)$ representa um ponto em $\R^2$, se quisermos interpolar esses dados podemos construir uma curva que passe por esses pontos fazendo da seguinte forma:

Mandamos o vetor de dados para $R_{x,d}$ e recebemos um polinômio que melhor aproxima eles.

Temos que esse polinômio aplicado ao avaliador $L_X$ é o mais próximo possível de $b$ na norma euclidiana. Vamos tomar outro conjunto de pontos e aplicar $L_Y$ ao polinômio ajustado. Isso nos dá o vetor:

$$
(p(y_1),\dots,p(y_n)).
$$

Isto é, realizamos $n$ novas amostragens da curva $p(x)$ através de $L_y$, nos dando o gráfico da curva nos pontos $(y_i, p(y_i))$.

A composta $L_Y \circ R_{x,d}$ pega os dados observados nos pontos $X$, ajusta um polinômio de grau no máximo $d$ por mínimos quadrados e depois avalia esse polinômio em novos pontos $Y$, fazendo uma reamostragem do polinômio em novos pontos, "completando" a curva.

Quando $m = d+1$, o polinômio ajustado vai passar exatamente pelos $m$ pontos, o que caracteriza uma interpolação. Porém, quando $m > d+1$, existem mais pontos do que coeficientes, o que torna impossível de existir uma interpolação para o conjunto de dados $X$ (nesse caso, fazemos uma regressão). 

### d)

Escolha alguns valores de $m$, $n$ e $d$ e calcule os valores singulares da composta, usando como intermediárias tanto a base canônica quanto a base dos polinômios interpoladores de Lagrange.

Qual a diferença entre as matrizes calculadas?

Na base canônica, temos que a representação matricial da composta vai ser:
$$
[L_Y \circ R_{x,d}]
=
V_yV_x^+
$$

onde $V_y$ é a matriz de Vandermonde associada aos pontos de $Y$ e $V_x^+$ é pseudoinversa da matriz de Vandermonde associada aos pontos de $X$. 

Como vimos na 3(a), o posto da composta vai ser $\operatorname(min){n,d+1}$. Sabemos também que matematicamente:

$$\operatorname{rank}(C) = \text{número de valores singulares não nulos de C}$$

Só que, numericamente, um valor singular nulo nem sempre vai ser realmente nulo, i.e., vamos analisar se numericamente isso acontece mesmo para diferentes valores de $n$, $d$ e $m$.

Faremos a mesma análise para a base de Lagrange. Vimos que matematicamente, o resultado da composta independe da base tomada para $P_d$, então o resultado esperado é o mesmo resultado que obtivemos no teste anterior. Porém, numericamente, Se as matrizes $V_x$ e $A_x$ são muito mal condicionadas, as duas pseudoinversas numéricas podem ser bem diferentes da pseudoinversa exata, o que faz com que a igualdade começe a quebrar.

```python
import numpy as np
import matplotlib.pyplot as plt


def transformar_em_lista(x):
    if isinstance(x, (list, tuple, np.ndarray)):
        return list(x)
    return [x]


def matriz_avaliacao_lagrange(pontos, nodes):
    """
    Matriz A tal que A[k,i] = ell_i(pontos[k]),
    usando a fórmula direta dos polinômios de Lagrange.
    """
    pontos = np.asarray(pontos, dtype=float)
    nodes = np.asarray(nodes, dtype=float)

    n = len(pontos)
    r = len(nodes)

    A = np.ones((n, r))

    for i in range(r):
        zi = nodes[i]

        for j in range(r):
            if j != i:
                zj = nodes[j]
                A[:, i] *= (pontos - zj) / (zi - zj)

    return A


def composta_canonica(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    Vx = np.vander(X, N=d+1, increasing=True)
    Vy = np.vander(Y, N=d+1, increasing=True)

    C = Vy @ np.linalg.pinv(Vx, rcond=rcond)

    return C, Vx


def composta_lagrange(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    # Nós usados para construir a base de Lagrange de P_d
    nodes = np.linspace(intervalo_X[0], intervalo_X[1], d+1)

    Ax = matriz_avaliacao_lagrange(X, nodes)
    Ay = matriz_avaliacao_lagrange(Y, nodes)

    C = Ay @ np.linalg.pinv(Ax, rcond=rcond)

    return C


def valores_singulares(A):
    return np.linalg.svd(A, compute_uv=False)


def plotar_valores_singulares_composta(
    m,
    n,
    d,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    rcond=1e-12,
    piso=1e-18,
    mostrar_resumo=True
):
    """
    Plota os valores singulares de L_Y o R_{X,d}
    comparando base canônica e base de Lagrange.

    m, n e d podem ser números ou listas.
    """
    valores_m = transformar_em_lista(m)
    valores_n = transformar_em_lista(n)
    valores_d = transformar_em_lista(d)

    casos = []

    for mi in valores_m:
        for ni in valores_n:
            for di in valores_d:
                if mi >= di + 1:
                    casos.append((mi, ni, di))
                else:
                    print(f"Ignorando caso inválido: m={mi}, n={ni}, d={di}, pois m < d+1.")

    if len(casos) == 0:
        print("Nenhum caso válido para plotar.")
        return

    if len(casos) > 12:
        print(f"Atenção: você está plotando {len(casos)} curvas em cada gráfico. Pode ficar visualmente poluído.")

    fig, axes = plt.subplots(1, 2, figsize=(15, 6), sharey=True)

    ax_can, ax_lag = axes

    for mi, ni, di in casos:
        r_teorico = min(ni, di + 1)

        # Base canônica
        try:
            C_can = composta_canonica(
                mi, ni, di,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            sigma_can = valores_singulares(C_can)
            sigma_can_plot = np.maximum(sigma_can, piso)

            ax_can.semilogy(
                range(1, len(sigma_can_plot) + 1),
                sigma_can_plot,
                marker="o",
                markersize=3,
                label=f"m={mi}, n={ni}, d={di}, r={r_teorico}"
            )

        except Exception as e:
            print(f"Erro na base canônica para m={mi}, n={ni}, d={di}: {e}")

        # Base de Lagrange
        try:
            C_lag = composta_lagrange(
                mi, ni, di,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            sigma_lag = valores_singulares(C_lag)
            sigma_lag_plot = np.maximum(sigma_lag, piso)

            ax_lag.semilogy(
                range(1, len(sigma_lag_plot) + 1),
                sigma_lag_plot,
                marker="o",
                markersize=3,
                label=f"m={mi}, n={ni}, d={di}, r={r_teorico}"
            )

        except Exception as e:
            print(f"Erro na base de Lagrange para m={mi}, n={ni}, d={di}: {e}")

        if mostrar_resumo:
            print(f"Caso m={mi}, n={ni}, d={di}: posto teórico esperado = min(n,d+1) = {r_teorico}")

    ax_can.set_title("Base canônica")
    ax_can.set_xlabel("Índice do valor singular")
    ax_can.set_ylabel("Valor singular")
    ax_can.grid(True, which="both", alpha=0.3)
    ax_can.legend(fontsize=8)

    ax_lag.set_title("Base de Lagrange")
    ax_lag.set_xlabel("Índice do valor singular")
    ax_lag.grid(True, which="both", alpha=0.3)
    ax_lag.legend(fontsize=8)

    fig.suptitle(
        f"Valores singulares de $L_Y\\circ R_{{X,d}}$\n",
        fontsize=13
    )

    plt.tight_layout()
    plt.show()
```

```python

plotar_valores_singulares_composta(
    m=80,
    d=50,
    n=[40, 80, 100]
)

plotar_valores_singulares_composta(
    m=150,
    d=100,
    n=[40, 80, 90]
)

plotar_valores_singulares_composta(
    m=80,
    d=50,
    n=[51, 80, 100]
)
```

A priori, temos alguns problemas na análise dos valores singulares:

Para calcularmos a composta, precisamos calcular a pseudoinversa da matriz de $L_X$ nas duas bases. Porém, isso pode ser acompanhado de alguns problemas:

1. A função np.linalg.pinv descarta valores singulares a partir de uma certa tolerância, o que faz com que pareça que só alguns valores singulares sobrevivam, e depois tudo cai pra perto de zero.

No gráfico, podemos ver uma queda brusca nos valores singulares em algum momento. Idealmente, isso deveria ocorrer quando $r$ = $\operatorname{min}{n,d+1}$. Entretanto, ao calcular $V_x^+$ e $A_x^+$ por meio do np.linalg.pinv, os valores singulares muito pequenos são tratados como zero (como discutimos anteriormente). Isso faz com que a curva caia muito antes do valor matemático do posto, mostrando uma instabilidade numérica.

Isso fica nítido no 3o gráfico, onde em ambos os casos a queda vai ocorrer antes do posto teórico.

2. O condicionamento das matrizes interfere no cálculo da composta. Se as matrizes $V_x$ e $A_x$ estão muito mal condicionadas, o cálculo da composta fica instável, dado que estamos construindo a pseudoinversa de matrizes mal condicionadas, amplificando mais ainda os erros que decorrem do mal condicionamento. 

Vimos que matematicamente, o resultado da composta independe da base tomada para $P_d$, então o resultado esperado é o mesmo resultado que obtivemos no teste anterior. Porém, numericamente, vimos que as matrizes $V_x$ e $A_x$ são muito mal condicionadas, as duas pseudoinversas numéricas podem ser bem diferentes da pseudoinversa exata, o que faz com que a igualdade começe a quebrar. 

Esse resultado fica mais nítido no caso de $d = 100$, pois a composta na base de Lagrange possui seus valores singulares muito próximos que zero antes quando comparados com a base canônica. Isso decorre do fato da matriz $A_x$ na base de Lagrange ficar muito mal condicionada para um valor tão grande de $d$ mais rápido do que na base canônica, como vimos no item 2. 

Assim, não podemos confiar no resultado dos valores singulares para esses casos extremos, e muito menos esperar NUMERICAMENTE que valha que o cálculo da composta independa da base.

Na próxima célula, vamos calcular a diferença entre a composta na base canônica e na base de Lagrange.


```python
def cond_svd_sem_corte(A):
    sigma = np.linalg.svd(A, compute_uv=False)

    if sigma[-1] == 0:
        return np.inf

    return sigma[0] / sigma[-1]


def grafico_diferenca_bases(
    m=150,
    n=90,
    d_max=120,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    rcond=1e-12
):
    valores_d = list(range(1, min(d_max, m-1) + 1))

    dif_max = []
    dif_rel = []
    conds_vandermonde = []
    conds_lagrange = []

    for d in valores_d:
        try:
            C_can, Vx = composta_canonica(
                m, n, d,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            C_lag, Ax = composta_lagrange(
                m, n, d,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            diff = C_can - C_lag

            # diferença absoluta máxima
            erro_max = np.max(np.abs(diff))

            # diferença relativa usando norma de Frobenius
            erro_rel = (
                np.linalg.norm(diff, ord="fro") /
                np.linalg.norm(C_can, ord="fro")
            )

            dif_max.append(erro_max)
            dif_rel.append(erro_rel)

            conds_vandermonde.append(cond_svd_sem_corte(Vx))
            conds_lagrange.append(cond_svd_sem_corte(Ax))

        except Exception:
            dif_max.append(np.nan)
            dif_rel.append(np.nan)
            conds_vandermonde.append(np.nan)
            conds_lagrange.append(np.nan)

    fig, axes = plt.subplots(1, 2, figsize=(15, 5))

    # Gráfico 1: diferença entre as compostas
    axes[0].semilogy(
        valores_d,
        dif_max,
        label=r"$\max |C_{\mathrm{can}}-C_{\mathrm{Lag}}|$",
        marker="o",
        markersize=3
    )

    axes[0].semilogy(
        valores_d,
        dif_rel,
        label="erro relativo",
        linestyle="--"
    )

    axes[0].set_xlabel("Grau do polinômio d")
    axes[0].set_ylabel("Diferença entre as compostas")
    axes[0].set_title("Diferença entre as compostas")
    axes[0].grid(True, which="both", alpha=0.3)
    axes[0].legend()

    # Gráfico 2: condicionamento das matrizes de avaliação
    axes[1].semilogy(
        valores_d,
        conds_vandermonde,
        label=r"$\kappa(V_X)$ — base canônica",
        marker="o",
        markersize=3
    )

    axes[1].semilogy(
        valores_d,
        conds_lagrange,
        label=r"$\kappa(A_X)$ — base de Lagrange",
        linestyle="--"
    )

    axes[1].set_xlabel("Grau do polinômio d")
    axes[1].set_ylabel("Condicionamento")
    axes[1].set_title("Condicionamento das matrizes de avaliação")
    axes[1].grid(True, which="both", alpha=0.3)
    axes[1].legend()

    fig.suptitle(
        f"Instabilidade numérica da composta para m={m}, n={n}, "
        f"X⊂[{intervalo_X[0]}, {intervalo_X[1]}]",
        fontsize=13
    )

    plt.tight_layout()
    plt.show()

    return valores_d, dif_max, dif_rel, conds_vandermonde, conds_lagrange
```

```python
grafico_diferenca_bases(
    m=150,
    n=90,
    d_max=140,
    intervalo_X=(-1, 1)
);
```

Em teoria, a composta $L_Y\circ R_{X,d}$ não depende da base escolhida para representar $P_d$, como falamos sobre anteriormente. Assim, as matrizes

$$
C_{\mathrm{can}}=V_YV_X^+
$$

e

$$
C_{\mathrm{Lag}}=A_YA_X^+
$$

deveriam coincidir. No gráfico da esquerda, observamos que essa igualdade é bem preservada para graus pequenos, já que a diferença entre as duas compostas é próxima de zero.

Porém, quandp o grau $d$ aumenta, a diferença entre as duas matrizes cresce bem mais rápido. Isso nos diz que a independência da base, mesmo que seja verdadeira matematicamnte, numericamente não vale quando as matrizes já estão muito mal condicionadas. No gráfico da direita, as matrizes usadas para calcular as pseudoinversas se tornam extremamente mal condicionadas, especialmente para graus altos.

Assim, diferenças de arredondamento na construção de $V_X$, $A_X$ e no cálculo das pseudoinversas são amplificadas, fazendo com que as duas formas de representar a mesma composta deixem de coincidir numericamente.

### e)

Qual o condicionamento da composta

$$
L_Y\circ R_{X,d}?
$$

Você pode fazer $n\to\infty$ sem prejudicar o condicionamento?

E $m\to\infty$ e $d\to\infty$?

Na base canônica, temos que a representação matricial da composta vai ser:
$$
[L_Y \circ R_{x,d}]
=
V_yV_x^+
$$

onde $V_y$ é a matriz de Vandermonde associada aos pontos de $Y$ e $V_x^+$ é pseudoinversa da matriz de Vandermonde associada aos pontos de $X$. 

Só que como temos 3 parâmetros, não podemos visualizar a variação de todos em um gráfico 2D. Assim, iremos fazer diferentes gráficos:

1. Fixando $m$ e variando $n$, $d$: se temos uma quantidade fixa de pontos em $X$, como o condicionamento de reamostrar o polinômio muda se variamos a quantidade de pontos ou o grau do polinômio. 

2. Fixando $m$, $n$ e variando $d$: vamos visualizar o efeito do grau no condicionamento


Note primeiramente que a matriz da composta de dimensões $n\times m$ é retangular, de posto $\operatorname{rank}=\operatorname{min}{n, d+1}$. Temos que a matriz pode te um núcleo não trivial, isto é, pode existir um $\sigma_i = 0$ associado ao núcleo, o que faz com o que o condicionamento da composta seja $\kappa(C) = \inf$. 

Assim, não faz sentido analisarmos o condicionamento usual, já que o condicionamento pode ser infinito, mesmo que a matriz tenha um comportamento bom nas direções dos valores singulares não nulos. Computacionalmente, vamos analisar entre as direções que a transformação não zera, quão diferente é o maior esticamento do menor esticamento. 

De forma que a gente possa fazer isso, vamos adaptar a nossa função de calcular condicionamentos, tomando o menor valor singular como o menor valor singular não nulo.

```python


def cond_efetivo(A, tol=1e-12):
    sigma = np.linalg.svd(A, compute_uv=False)

    sigma_nz = sigma[sigma > tol * sigma[0]]

    if len(sigma_nz) == 0:
        return np.inf

    return sigma_nz[0] / sigma_nz[-1]



def heatmap_cond_composta_m_fixo(
    m,
    n_min=2,
    n_max=150,
    passo_n=5,
    d_min=1,
    d_max=None,
    passo_d=2,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    tol=1e-12,
    rcond=1e-12,
    log_vmin=0,
    log_vmax=16
):
    if d_max is None:
        d_max = m - 1

    d_max = min(d_max, m - 1)

    valores_n = list(range(n_min, n_max + 1, passo_n))
    valores_d = list(range(d_min, d_max + 1, passo_d))

    grade_can = np.full((len(valores_n), len(valores_d)), np.nan)
    grade_lag = np.full((len(valores_n), len(valores_d)), np.nan)

    X = np.linspace(intervalo_X[0], intervalo_X[1], m)

    for idx_d, d in enumerate(valores_d):

        # Matriz de X na base canônica
        try:
            Vx = np.vander(X, N=d + 1, increasing=True)
            Vx_pinv = np.linalg.pinv(Vx, rcond=rcond)
        except Exception:
            Vx_pinv = None

        # Matriz de X na base de Lagrange
        try:
            nodes = np.linspace(intervalo_X[0], intervalo_X[1], d + 1)
            Ax = matriz_avaliacao_lagrange(X, nodes)
            Ax_pinv = np.linalg.pinv(Ax, rcond=rcond)
        except Exception:
            Ax_pinv = None

        for idx_n, n in enumerate(valores_n):
            Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

            # Base canônica
            if Vx_pinv is not None:
                try:
                    Vy = np.vander(Y, N=d + 1, increasing=True)
                    C_can = Vy @ Vx_pinv

                    c = cond_efetivo(C_can, tol=tol)

                    if np.isfinite(c) and c > 0:
                        grade_can[idx_n, idx_d] = np.log10(c)
                    else:
                        grade_can[idx_n, idx_d] = log_vmax

                except Exception:
                    grade_can[idx_n, idx_d] = np.nan

            # Base de Lagrange
            if Ax_pinv is not None:
                try:
                    Ay = matriz_avaliacao_lagrange(Y, nodes)
                    C_lag = Ay @ Ax_pinv

                    c = cond_efetivo(C_lag, tol=tol)

                    if np.isfinite(c) and c > 0:
                        grade_lag[idx_n, idx_d] = np.log10(c)
                    else:
                        grade_lag[idx_n, idx_d] = log_vmax

                except Exception:
                    grade_lag[idx_n, idx_d] = np.nan

    # Truncamento visual
    grade_can = np.clip(grade_can, log_vmin, log_vmax)
    grade_lag = np.clip(grade_lag, log_vmin, log_vmax)

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(16, 6),
        sharey=True,
        constrained_layout=True
    )

    cmap = plt.cm.viridis.copy()
    cmap.set_bad(color="white")

    im0 = axes[0].imshow(
        grade_can,
        extent=[valores_d[0], valores_d[-1], valores_n[0], valores_n[-1]],
        origin="lower",
        cmap=cmap,
        vmin=log_vmin,
        vmax=log_vmax,
        aspect="auto",
        interpolation="nearest"
    )

    axes[0].set_title("Base canônica")
    axes[0].set_xlabel("Grau do polinômio d")
    axes[0].set_ylabel("Número de pontos n em Y")

    im1 = axes[1].imshow(
        grade_lag,
        extent=[valores_d[0], valores_d[-1], valores_n[0], valores_n[-1]],
        origin="lower",
        cmap=cmap,
        vmin=log_vmin,
        vmax=log_vmax,
        aspect="auto",
        interpolation="nearest"
    )

    axes[1].set_title("Base de Lagrange")
    axes[1].set_xlabel("Grau do polinômio d")

    fig.colorbar(
        im1,
        ax=axes,
        label=r"$\log_{10}(\kappa_{\mathrm{eff}}(L_Y\circ R_{X,d}))$",
        shrink=0.9
    )

    fig.suptitle(
        f"Condicionamento efetivo da composta $L_Y\\circ R_{{X,d}}$ com m={m}\n"
        f"$X\\subset[{intervalo_X[0]}, {intervalo_X[1]}]$, "
        f"$Y\\subset[{intervalo_Y[0]}, {intervalo_Y[1]}]$",
        fontsize=13
    )

    plt.show()

    return valores_n, valores_d, grade_can, grade_lag
```

Nos mapas de calor, fixamos $m=100$ e variamos o valor de $n$ e o grau $d$. A quantidade que representamos é

$$
\log_{10}\left(\kappa(L_Y\circ R_{X,d})\right),
$$

isto é, o log do condicionamento alterado (que considera o menor valor singular não nulo) da composta.

Podemos observar uma mudança de comportamento próximo de $n=d+1$. Isso é esperado, pois

$$
\operatorname{rank}(L_Y\circ R_{X,d})=\min\{n,d+1\}.
$$

Assim, quando $n<d+1$, o posto da composta vai ser limitado ao núm. de pontos em $Y$ (vimos que numericamente, o posto não necessariamente vai ser o mínimo, dado que alguns valores singulares muito próximos do erro de precisão da máquina são considerados zero), mas quando $n>d+1$, o posto vai a ser a dimensão do espaço $P_d$.

Na base canônica, o crescimento do condicionamento é mais "comportado" e está bastante associado ao aumento do grau $d$. Isso está de acordo com o comportamento que vimos das matrizes de Vandermonde, que se tornam mais mal condicionadas quando o grau cresce.

Na base de Lagrange, o comportamento é parecido pra graus baixos e moderados, mas aparecem irregularidades pra graus mais altos. Essas irregularidades não vão contradizer a teoria de que a composta independe da base. Elas indicam instabilidade numérica na construção da matriz na base de Lagrange e no cálculo da pseudoinversa $A_X^+$. Em particular, como estamos usando a fórmula direta dos polinômios de Lagrange, os produtórios envolvidos podem acumular erros numéricos quando $d$ cresce.

Comparando os resultados pra $m=50$ e $m=100$, observamos que o padrão se mantém parecido. Nos dois casos, o crescimento do grau $d$ é o principal fator associado ao aumento do condicionamento. Porém, aumentar $m$ muda a região válida do gráfico, já que a condição $m\geq d+1$ nso deixa considerar graus maiores. Além disso, como $R_{X,d}$ depende da pseudoinversa da matriz em $X$, mudar $m$ também muda o problema de mínimos quadrados original. Assim, $m$ não é irrelevante, já que ele muda a matriz $V_X$ ou $A_X$ e, também, a pseudoinversa usada na composta.

Assim, os gráficos confirmam que o grau $d$ é o parâmetro mais sensível. O número de pontos $m$ influencia, mas o aumento de $d$ tem um impacto muito maior no condicionamento.

```python
def comparar_cond_composta_d_variando(
    m,
    n,
    d_min=1,
    d_max=None,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    tol=1e-12,
    rcond=1e-12,
    teto=1e50
):

    if d_max is None:
        d_max = m - 1

    # Pela hipótese do problema, precisamos de m >= d+1.
    d_max = min(d_max, m - 1)

    valores_d = list(range(d_min, d_max + 1))

    conds_can = []
    conds_lag = []

    for d in valores_d:

        # Base canônica
        try:
            C_can = composta_canonica(
                m=m,
                n=n,
                d=d,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            c_can = cond_efetivo(C_can, tol=tol)

            if not np.isfinite(c_can):
                c_can = teto

            conds_can.append(c_can)

        except Exception:
            conds_can.append(np.nan)

        # Base de Lagrange
        try:
            C_lag = composta_lagrange(
                m=m,
                n=n,
                d=d,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            c_lag = cond_efetivo(C_lag, tol=tol)

            if not np.isfinite(c_lag):
                c_lag = teto

            conds_lag.append(c_lag)

        except Exception:
            conds_lag.append(np.nan)

    plt.figure(figsize=(9, 6))

    plt.semilogy(
        valores_d,
        conds_can,
        label="Base canônica",
        marker="o",
        markersize=3
    )

    plt.semilogy(
        valores_d,
        conds_lag,
        label="Base de Lagrange",
        marker="o",
        markersize=3,
        linestyle="--"
    )

    plt.xlabel("Grau do polinômio d")
    plt.ylabel(r"$\kappa_{\mathrm{eff}}(L_Y\circ R_{X,d})$")
    plt.title(
        f"Condicionamento efetivo da composta com m={m}, n={n}\n"
        f"$X\\subset[{intervalo_X[0]}, {intervalo_X[1]}]$, "
        f"$Y\\subset[{intervalo_Y[0]}, {intervalo_Y[1]}]$"
    )

    plt.grid(True, which="both", alpha=0.3)
    plt.legend()
    plt.show()

    return valores_d, conds_can, conds_lag
```

```python
comparar_cond_composta_d_variando(m=150, n=80, d_max=100);
comparar_cond_composta_d_variando(m=100, n=80, d_max=100);
comparar_cond_composta_d_variando(m=150, n=150, d_max=100);
```

Fixando $m$ e $n$ e variando o grau $d$, observamos que o condicionamento da composta se mantém baixo pra graus pequenos, mas acaba crescendo muito rápido a partir de certo valor de $d$. Esse comportamento nos diz que o aumento do grau é o principal responsável pela perda de estabilidade numérica.

Em todos os gráficos, a base canônica tem esse salto mais cedo, por volta de $d\approx 17$ ou $d\approx 18$. Na base de Lagrange, ela se mantém bem condicionada por mais tempo, onde o salto só acontece pra graus maiores. Assim, nesses testes, a base de Lagrange parece "adiar" o início do mau condicionamento da composta.

Depois do salto, as duas bases nos mostram condicionamentos muito altos, da ordem de $10^{11}$ ou $10^{12}$. Assim, pra valores de $d$, a composta acaba se tornando numericamente sensível independentemente da base utilizada, como vimos anteriormente.

As quedas bruscas que podemos ver na curva da base de Lagrange pra graus altos devem ser interpretadas com cuidado. Pra valores grandes de $d$, os nós de Lagrange ficam muito próximos entre si (no linspace), o que torna os produtórios numericamente instáveis. Logo, essas quedas não necessariamente significam uma melhora de verdade da estabilidade.

Finalmente, vamos ver o que acontece quando levamos os parâmetros pro infinito (vamos analisar o comportamento de como eles crescem, já que calcular eles no infinito é computacionalmente inviável).

Assim, vamos analisar diferentes casos, com $d$ fixo com $m,n$ crescentes, $d$ proporcional a $m$, e o caso em que $d=m-1$.

```python
import numpy as np
import matplotlib.pyplot as plt


def valores_singulares_produto_ba(B, A, tol=1e-12):

    M = A @ A.T          # r x r
    G = B.T @ B          # r x r

    # Matriz simétrica equivalente:
    # M^{1/2} G M^{1/2}
    eig_M, Q = np.linalg.eigh(M)
    eig_M = np.maximum(eig_M, 0)

    sqrt_M = Q @ np.diag(np.sqrt(eig_M)) @ Q.T

    S = sqrt_M @ G @ sqrt_M
    S = (S + S.T) / 2   # força simetria numérica

    eig_S = np.linalg.eigvalsh(S)
    eig_S = np.maximum(eig_S, 0)

    sigma = np.sqrt(eig_S)
    sigma = np.sort(sigma)[::-1]

    # remove valores numericamente nulos
    if sigma[0] == 0:
        return np.array([])

    sigma_nz = sigma[sigma > tol * sigma[0]]

    return sigma_nz

def fatores_composta_canonica(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    Vx = np.vander(X, N=d+1, increasing=True)
    Vy = np.vander(Y, N=d+1, increasing=True)

    Vx_pinv = np.linalg.pinv(Vx, rcond=rcond)

    # C = Vy @ Vx_pinv
    return Vy, Vx_pinv


def fatores_composta_lagrange(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    nodes = np.linspace(intervalo_X[0], intervalo_X[1], d+1)

    Ax = matriz_avaliacao_lagrange(X, nodes)
    Ay = matriz_avaliacao_lagrange(Y, nodes)

    Ax_pinv = np.linalg.pinv(Ax, rcond=rcond)

    # C = Ay @ Ax_pinv
    return Ay, Ax_pinv


def cond_efetivo_produto_ba(B, A, tol=1e-12):
    sigma = valores_singulares_produto_ba(B, A, tol=tol)

    if len(sigma) == 0:
        return np.inf

    return sigma[0] / sigma[-1]


def experimento_assintotico(
    valores_m,
    regra_n,
    regra_d,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    tol=1e-12,
    rcond=1e-12,
    teto=1e50,
    calcular_lagrange=True
):
    conds_can = []
    conds_lag = []

    valores_n = []
    valores_d = []

    for m in valores_m:
        n = regra_n(m)
        d = regra_d(m)

        valores_n.append(n)
        valores_d.append(d)

        if m < d + 1:
            conds_can.append(np.nan)
            conds_lag.append(np.nan)
            continue

        # Base canônica
        try:
            B_can, A_can = fatores_composta_canonica(
                m=m,
                n=n,
                d=d,
                intervalo_X=intervalo_X,
                intervalo_Y=intervalo_Y,
                rcond=rcond
            )

            c_can = cond_efetivo_produto_ba(B_can, A_can, tol=tol)

            if not np.isfinite(c_can):
                c_can = teto

            conds_can.append(c_can)

        except Exception as e:
            print(f"Erro canônica em m={m}, n={n}, d={d}: {e}")
            conds_can.append(np.nan)

        # Base de Lagrange
        if calcular_lagrange:
            try:
                B_lag, A_lag = fatores_composta_lagrange(
                    m=m,
                    n=n,
                    d=d,
                    intervalo_X=intervalo_X,
                    intervalo_Y=intervalo_Y,
                    rcond=rcond
                )

                c_lag = cond_efetivo_produto_ba(B_lag, A_lag, tol=tol)

                if not np.isfinite(c_lag):
                    c_lag = teto

                conds_lag.append(c_lag)

            except Exception as e:
                print(f"Erro Lagrange em m={m}, n={n}, d={d}: {e}")
                conds_lag.append(np.nan)
        else:
            conds_lag.append(np.nan)

    return {
        "m": valores_m,
        "n": valores_n,
        "d": valores_d,
        "cond_can": conds_can,
        "cond_lag": conds_lag
    }

def plotar_experimento_assintotico(resultado, titulo):
    valores_m = resultado["m"]
    valores_n = resultado["n"]
    valores_d = resultado["d"]
    cond_can = resultado["cond_can"]
    cond_lag = resultado["cond_lag"]

    plt.figure(figsize=(9, 6))

    plt.semilogy(
        valores_m,
        cond_can,
        marker="o",
        markersize=4,
        label="Base canônica"
    )

    if not np.all(np.isnan(cond_lag)):
        plt.semilogy(
            valores_m,
            cond_lag,
            marker="o",
            markersize=4,
            linestyle="--",
            label="Base de Lagrange"
        )

    plt.xlabel("m")
    plt.ylabel(r"$\kappa_{\mathrm{eff}}(L_Y\circ R_{X,d})$")
    plt.title(titulo)
    plt.grid(True, which="both", alpha=0.3)
    plt.legend()

    plt.show()

    print("Tabela dos parâmetros:")
    for mi, ni, di, cc, cl in zip(valores_m, valores_n, valores_d, cond_can, cond_lag):
        print(f"m={mi:6d}, n={ni:6d}, d={di:6d}, cond_can={cc:.3e}, cond_lag={cl:.3e}")
```

```python
valores_m = [50, 100, 200, 500, 1000, 5000, 10000, 50000, 100000]

resultado_d_fixo = experimento_assintotico(
    valores_m=valores_m,
    regra_n=lambda m: 2*m,
    regra_d=lambda m: 10,
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    calcular_lagrange=True
)

plotar_experimento_assintotico(
    resultado_d_fixo,
    r"Comportamento assintótico: $d=10$ fixo, $n=2m$"
)
```

Nesse experimento, vamos manter o grau fixo em $d=10$ e fazemos $m\to\infty$, tomando $n=2m$. Assim, a dimensão de $P_d$ se mantém fixa:

$$
\dim(P_d)=11.
$$

Podemos ver que o condicionamento se estabiliza próximo de $1$ conforme $m$ cresce. Isso nos diz que, quando o grau se mantém fixo, aumentar a quantidade de pontos de amostragem em $X$ e $Y$ não piora muito o condicionamento da composta.

Esse comportamento contrasta bastante com os experimentos que fizemos antes, nos quais o aumento de $d$ causava um crescimento muito grande do condicionamento. Assim, o experimento sugere que o grau $d$, e não apenas o tamanho de $m$ e $n$, é o principal fator responsável pela instabilidade numérica (como nós falamos sobre antes).

```python
valores_m = list(range(20, 301, 20))

resultado_d_prop = experimento_assintotico(
    valores_m=list(range(20, 301, 20)),
    regra_n=lambda m: 2*m,
    regra_d=lambda m: int(0.3*m),
    intervalo_X=(-1, 1),
    intervalo_Y=(-1, 1),
    calcular_lagrange=True
)

plotar_experimento_assintotico(
    resultado_d_prop,
    r"Comportamento assintótico: $d=\lfloor 0.3m\rfloor$, $n=2m$"
)
```

Neste experimento, tomamos

$$
n=2m
\qquad\text{e}\qquad
d=\lfloor 0.3m\rfloor.
$$

Assim, quando $m$ cresce, aumentamos a quantidade de pontos em $X$ e $Y$ e também aumentamos a dimensão de $P_d$.

Diferente do caso em que $d$ foi mantido fixo, aqui o condicionamento cresce mais rápido. Isso indica que o fator decisivo pra instabilidade não é só o aumento de $m$ e $n$, mas principalmente o crescimento do grau $d$ (como nós já vimos antes).

Na base canônica, o condicionamento começa a crescer mais cedo, o que nos mostra a sensibilidade da matriz de Vandermonde com aumento do grau, algo que esperamos. Já na base de Lagrange, o condicionamento se mantém baixo por mais tempo, mas depois cresce de forma muito mais abrupta pra graus altos. Esse crescimento está associado à instabilidade numérica da construção da matriz na base de Lagrange e ao cálculo da pseudoinversa, como falamos sobre antes.

Assim, o experimento mostra que o comportamento assintótico da composta depende fortemente da relação entre $d$ e $m$. Quando $d$ é fixo, o condicionamento permanece controlado; quando $d$ cresce proporcionalmente a $m$, o condicionamento aumenta de forma bem mais rápida.

```python
valores_m = list(range(10, 151, 10))

resultado_quase_interp = experimento_assintotico(
    valores_m=valores_m,
    regra_n=lambda m: m,
    regra_d=lambda m: m - 1,
    calcular_lagrange=True
)

plotar_experimento_assintotico(
    resultado_quase_interp,
    r"Regime quase interpolatório: $d=m-1$, $n=m$"
)
```

No caso em que $d=m-1$ e $n=m$, estamos no caso interpolatório, já que $d+1=m$. Além disso, como tomamos $X=Y$, a composta estudada é

$$
L_X\circ R_{X,d}.
$$

Na base de Lagrange, usando como nós os próprios pontos de $X$, a matriz é a identidade:

$$
A_X=I_m.
$$

Como $Y=X$, também temos $A_Y=I_m$. Portanto,

$$
C_{\mathrm{Lag}}=A_YA_X^+=I_m,
$$

e, assim,

$$
\kappa_(C_{\mathrm{Lag}})=1.
$$

Isso explica a curva constante da base de Lagrange.

Na base canônica, em aritmética exata, também deveríamos obter

$$
C_{\mathrm{can}}=V_XV_X^+=I_m,
$$

pois $V_X$ é quadrada e inversível. Porém, o gráfico mostra que o condicionamento calculado cresce muito rápido. Isso acontece porque a matriz de Vandermonde fica extremamente mal condicionada quando o grau cresce (como vimos antes). Assim, a pseudoinversa numérica calculada pelo computador não vai reproduzir exatamente a inversa teórica.

Assim, esse experimento não contradiz a teoria que vimos antes. Pelo contrário, ele mostra que, no caso interpolatório, a base de Lagrange vai preservar numericamente a estrutura que esperamos da composta, enquanto a base canônica sofre com a instabilidade da matriz de Vandermonde.

---

A análise final nos diz que podemos fazer $n\to\infty$ sem prejudicar muito o condicionamento, desde que o grau $d$ permaneça fixo e os pontos de $Y$ estejam bem distribuídos no mesmo intervalo (usando o linspace). Isso acontece porque aumentar $n$ apenas aumenta a quantidade de avaliações do polinômio ajustado, mas não aumenta a dimensão de $P_d$.

Agora, fazer $m\to\infty$ com $d$ fixo também tende a ser estável, porque o espaço $P_d$ continua tendo dimensão fixa $d+1$. Nesse caso, aumentamos a quantidade de dados usados no ajuste do polinôimio, mas não aumentamos a complexidade do modelo.

Por outro lado, quando fazemos $m\to\infty$ e $d\to\infty$ ao mesmo tempo, o condicionamento pode piorar de forma muito rápida. Isso acontece porque a dimensão de $P_d$ também cresce, o que torna as matrizes de avaliação e suas pseudoinversas mais sensíveis numericamente. Os testes com $d=\lfloor 0.3m\rfloor$ mostram esse comportamento, já que o condicionamento cresce de forma significativa enquanto $m$ aumenta.

Assim, como vimos em diversos casos, o parâmetro mais delicado não é só o número de pontos $m$ ou $n$, mas principalmente o crescimento do grau $d$. Aumentar a quantidade de pontos é relativamente seguro quando $d$ é fixo, mas aumentar o grau junto com o tamanho do problema pode tornar a composta mal condicionada.

Pontos de Chebyshev foram criados com a interpolação polinomial em mente. Como vimos no item anterior, o polinômio interpolador tende a oscilar muito nos limites do intervalo isso se chama fenômeno de Runge e ocorre quando interpolamos pontos igualmente espaçados em um intervalo e o erro aumenta muito nas extremidades forçando nossa função a oscilar demais.
Considerando uma função $f(x)$ que une exatamente nossos pontos mas que não temos conhecimento de que função é essa, temos que o erro da nossa aproximação $p_{n-1}(x)$, onde $n-1$ é o grau do polinômio, para a função implícita no ponto de abcissa $x$ é:
$$
E(x) = f(x) - p_{n-1}(x) = K \prod_{i=1}^n (x-x_i)
$$
Onde K é uma constante que depende da função implícita desconhecida que não podemos fazer nada a respeito e para o exemplo podemos ignora-lo.
Podemos entender cada fator do produtório como a distância do $x$ desconhecido até o nó $x_i$. Quando perto das bordas, para valores de $x_i$ igualmente espaçados essa multiplicação explode.
Os pontos de Chebyshev são os valores de $x_i$ que minimizam:
$$
max_{x \in [-1,1]} \prod_{i=1}^n (x-x_i) 
$$

Podemos observar sua efetividade no seguinte exemplo:

```python

d = 20
m = 20
n = 200
x = np.arange(1,m+1)
x = np.cos((2*x - 1) * np.pi / (2*m) )
y = np.linspace(-1,1, n)
b = np.random.rand(m,1)

Vx = np.vander(x, N=d+1, increasing=True)
Vinv = np.linalg.pinv(Vx)

V = np.vander(y, N=d+1, increasing=True)

points = V @ Vinv @ b

plt.plot(y, points)
plt.scatter(x, b)
plt.ylim((-2, 2))
```

Aqui analisaremos alguns valores singulares da composta:

```python
def chebyshev_nodes(n_points, intervalo=(-1, 1)):
    """
    Gera n_points nós de Chebyshev mapeados para o intervalo (a, b).
    """
    a, b = intervalo
    i = np.arange(1, n_points + 1)
    nodes_11 = np.cos((2 * i - 1) / (2 * n_points) * np.pi)
    
    nodes_11 = np.flip(nodes_11)
    
    return 0.5 * (a + b) + 0.5 * (b - a) * nodes_11


def lagrange_eval(t, nodes, i):
    produto = 1.0
    zi = nodes[i]

    for j, zj in enumerate(nodes):
        if j != i:
            produto *= (t - zj) / (zi - zj)

    return produto


def matriz_avaliacao_lagrange(pontos, nodes):
    pontos = np.asarray(pontos, dtype=float)
    nodes = np.asarray(nodes, dtype=float)

    n = len(pontos)
    r = len(nodes)

    A = np.zeros((n, r))

    for k in range(n):
        for i in range(r):
            A[k, i] = lagrange_eval(pontos[k], nodes, i)

    return A


def composta_canonica(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    V_X = np.vander(X, N=d+1, increasing=True)
    V_Y = np.vander(Y, N=d+1, increasing=True)

    C = V_Y @ np.linalg.pinv(V_X, rcond=rcond)

    return C
    
def composta_lagrange(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    nodes = np.linspace(intervalo_X[0], intervalo_X[1], d+1)

    A_X = matriz_avaliacao_lagrange(X, nodes)
    A_Y = matriz_avaliacao_lagrange(Y, nodes)

    C = A_Y @ np.linalg.pinv(A_X, rcond=rcond)

    return C

def composta_canonica_chebyshev(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = chebyshev_nodes(m, intervalo_X)
    
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    V_X = np.vander(X, N=d+1, increasing=True)
    V_Y = np.vander(Y, N=d+1, increasing=True)

    C = V_Y @ np.linalg.pinv(V_X, rcond=rcond)

    return C


def composta_lagrange_chebyshev(m, n, d, intervalo_X=(-1, 1), intervalo_Y=(-1, 1), rcond=1e-12):
    X = np.linspace(intervalo_X[0], intervalo_X[1], m)
    Y = np.linspace(intervalo_Y[0], intervalo_Y[1], n)

    cnodes = chebyshev_nodes(d + 1, intervalo_X)
    nodes = np.linspace(intervalo_Y[0], intervalo_Y[1], d+1)


    A_X = matriz_avaliacao_lagrange(X, cnodes)
    A_Y = matriz_avaliacao_lagrange(Y, nodes)

    C = A_Y @ np.linalg.pinv(A_X, rcond=rcond)

    return C


d_vals = [20, 40, 60]
m_vals = [20, 60, 120]
n_vals = [40, 120, 240]

def plot_sin_vals(pontos_dist="cheb", base="canonica", intervalo_X=(-1, 1), intervalo_Y=(-1, 1)):
    fig, axs = plt.subplots(1, 3, figsize=(15, 4))
    for i, m in enumerate(m_vals):
        d = d_vals[i]
        n = n_vals[i]
        if pontos_dist == "cheb":
            if base == "canonica":
                C = composta_canonica_chebyshev(m, n, d, intervalo_X=intervalo_X, intervalo_Y=intervalo_Y)
            if base == "lagrange":
                C = composta_lagrange_chebyshev(m, n, d, intervalo_X=intervalo_X, intervalo_Y=intervalo_Y)
        if pontos_dist == "eq":
            if base == "canonica":
                C = composta_canonica(m, n, d, intervalo_X=intervalo_X, intervalo_Y=intervalo_Y)
            if base == "lagrange":
                C = composta_lagrange(m, n, d, intervalo_X=intervalo_X, intervalo_Y=intervalo_Y)
        s_vals = np.linalg.svd(C, compute_uv=False)
        axs[i].plot(s_vals)
        axs[i].set_yscale("log")
        fig.suptitle(f"Valores singulares ordenados para diferentes compostas (base: {base})")
        axs[i].set_title(f"m = {m}; n = {n}; d = {d}")
        axs[i].set_ylabel("valor singular")
        axs[i].set_xlabel("posição do valor singular")
    plt.tight_layout()
    plt.show()

plot_sin_vals()
plot_sin_vals(base="lagrange")
```

**ANÁLISE GRÁFICA**

Vemos que o descrescimento dos valores singulares ocorre de forma gradual até certo ponto onde há um descrescimento repentino para números que são praticamente zero. O fenômeno observado se dá proximo ao posto da matriz, que como vimos anteriormente é $min\{n, d\}$, esses valores próximos a zero deveriam ser zero já que a nossa svd tem forma:
$$
\Sigma = \begin{bmatrix} \hat{\Sigma} \\ 0 \end{bmatrix} 
$$
Concluímos que quanto mais próximos esses valores estão de 0 melhor a nossa representação da transformação composta. Observando os resultados podemos enxergar que as melhores aproximações seriam a segunda e a terceira utilizando a base de lagrange onde o erro numérico é na grandeza de $10^{-10}$.

```python

m_fixo = 100
n_fixo = 100
d_fixo = 15  

valores_d = list(range(1, m_fixo))
valores_m = list(range(d_fixo + 1, 150))  
valores_n = list(range(1, 2 * m_fixo))

conds_can = []
conds_lag = []
conds_m = []
conds_n = []

for d in valores_d:
    C_can = composta_canonica(m_fixo, n_fixo, d)
    C_lag = composta_lagrange_chebyshev(m_fixo, n_fixo, d)
    
    conds_can.append(cond_efetivo(C_can))
    conds_lag.append(cond_efetivo(C_lag))

for m_atual in valores_m:
    C_lag = composta_lagrange_chebyshev(m_atual, n_fixo, d_fixo)
    conds_m.append(cond_efetivo(C_lag))
    
for n_atual in valores_n:
    C_lag = composta_lagrange_chebyshev(m_fixo, n_atual, d_fixo)
    conds_n.append(cond_efetivo(C_lag))

fig, axs = plt.subplots(1, 3, figsize=(15, 5)) 
fig.suptitle(f"Condicionamento efetivo de $L_Y \\circ R_{{X,d}}$ com Nós de Chebyshev", fontsize=14)

axs[0].plot(valores_d, conds_can, label="Base canônica", marker="o", markersize=2)
axs[0].plot(valores_d, conds_lag, label="Base de Lagrange", linestyle="--")
axs[0].set_xlabel("Grau do polinômio d")
axs[0].set_ylabel("Condicionamento efetivo")
axs[0].legend()

axs[1].plot(valores_m, conds_m, color='orange')
axs[1].set_xlabel("Quantidade de pontos m (X)")
axs[1].set_ylabel("Condicionamento efetivo")
axs[1].set_title(f"m variando (d={d_fixo}, n={n_fixo})", fontsize=10)

axs[2].plot(valores_n, conds_n, color='green')
axs[2].set_xlabel("Quantidade de pontos n (Y)")
axs[2].set_ylabel("Condicionamento efetivo")
axs[2].set_title(f"n variando (d={d_fixo}, m={m_fixo})", fontsize=10)

for ax in axs:
    ax.set_yscale("log")
    ax.grid(True, which="both", alpha=0.3)

plt.tight_layout()
plt.show()
```

**ANÁLISE GRÁFICA:**

Observa-se no primeiro gráfico que o uso de nós de Chebychev quando usando com a base de Lagrange reduz drasticamente o condicionamento da composta comparado com pontos equidistantes na base canônica a medida que o grau do polinômio cresce. Em contraste com o resultado obtido em 3(d), mesmo para base de Lagrange que já possuia um melhor resultado, o condicionamento da composta usando pontos de Chebchev tem uma tolênrancia bem maior a polinômios de grau mais elevados, sendo assim temos resultados mais confiáveis para $L_Y \circ R_{X,d}$ para um intervalo maior de $d$. Entretanto, depois que o condicionamento estoura esse permanece semelhante ao condicionamento utilizando os pontos igualmente espaçados.
Respondendo as perguntas do enunciado, apenas o crescimento de m não afeta o condicionamento negativamente o que faz sentido visto que mais dados nos ajudaria a criar um interpolador melhor.

### g) Extrapolação 1

Considere

$$
X=\operatorname{linspace}(-1,1,m),
$$

e

$$
Y=\operatorname{linspace}(-1.5,1.5,n).
$$

O que acontece agora?


Quando realizamos uma extrapolação estamos tentando criar uma função na intenção de analizar pontos fora do intervalo conhecido, em contraposição com a interpolação que busca pontos entre os dados que já temos, valores intermediários. Nesse caso estamos buscando uma função que possa "prever" o futuro. Dessa forma, Podemos relacionar a composta com um modelo de regressão polinomial em batch (o modelo é treinado com todos os dados de uma só vez) onde se busca minimizar o erro entre os valores previstos e os dados reais. Isso muda a forma como enxergamos a situação visto que a matriz $V_x^+$, correspondente a $R_{X,d}$, segue exatamente a forma da pseudoinversa de Moore-Penrose:
$$
\Phi^+ = (\Phi^T \Phi) \Phi^T
$$
Onde $\Phi$ é a matriz cujo as colunas contém uma transformação não linear das nossas entradas $x_i$. No caso da regressão polinomial ela tem a forma:
$$
\Phi=\begin{bmatrix} 
x_1^0 & x_1^1 & x_1^2 \\
x_2^0 & x_2^1 & x_2^2 \\
x_3^0 & x_3^1 & x_3^2
\end{bmatrix}
$$ 
A matriz de Moore-Penrose é encontrada quando minimizamos a função de erro quadrado:
$$
E(p) = \sum_{k=1}^m (b_i - p^T \phi(x_i))
$$
Onde $p$ são os coeficientes do polinômio e $\phi(x_i)$ é a linha $i$ da matriz $\Phi$

Ademais, quando $y_i$ está em um intervalo maior que $x_i$ a operação $L_y$ mostra sua utilidade na vizualização do modelo treinado no passo anterior e como ele se comporta fora do intervalo dos dados de treinamento $X$.

```python

d = 5
m = 50
n = 200
x = np.linspace(-1,1, m)
y = np.linspace(-1.5,1.5, n)
b = np.random.rand(m,1)

Vx = np.vander(x, N=d+1, increasing=True)
Vinv = np.linalg.pinv(Vx)

V = np.vander(y, N=d+1, increasing=True)

points = V @ Vinv @ b

plt.plot(y, points)
plt.scatter(x, b)
plt.ylim((-2, 2))
```

O gráfico mostra os pontos $(x_i, b_i)$ onde a linha mostra a previsão de valores desconhecidos de $b_i$.
É interessante apontar que um ótimo interpolador é um péssimo extrapolador, já que o polinômio passa tão precisamente pelos dados de treinamento que ele começa a "prever o ruído" nos dados e perde a generalidade para dados novos. No mundo de machine learning chamamos esse fenômeno de overfitting.

```python
plot_sin_vals(pontos_dist="eq", base="canonica", intervalo_X=(-1, 1), intervalo_Y=(-1.5, 1.5))
plot_sin_vals(pontos_dist="eq", base="lagrange", intervalo_X=(-1, 1), intervalo_Y=(-1.5, 1.5))
```

Uma breve análise dos valores sigulares, como em 3(f), nos mostra que o problema da extrapolação é em geral sujeito a mais erros numéricos que a interpaloção mesmo para a base de Lagrange, posto que os menores valores singulares estão em sua maioria mais distantes do zero real.

### h)

Adianta usar os pontos de Chebyshev em $X$?


Para a função do extrapolador a utilização de pontos de Chebyshev não muda praticamente nada. A regressão polinomial é bem limitada visto que fora do intervalo $[-1,1]$ o nosso polinômio vai quase sempre pra mais ou menos infinito o que não necessariamente representa a tendência dos nossos pontos. Entretanto, como foi dito anteriormente os nós de Chebychev minimizam oscilações nas bordas do intervalo então conseguiremos uma função mais "comportada" para graus mais elevados o que em geral é bom. 

```python
plot_sin_vals(pontos_dist="cheb", base="canonica", intervalo_X=(-1, 1), intervalo_Y=(-1.5, 1.5))
plot_sin_vals(pontos_dist="cheb", base="lagrange", intervalo_X=(-1, 1), intervalo_Y=(-1.5, 1.5))
```

Mesmo os nós de Chebyshev não agregando muito para a extrapolação de forma superficial vemos a partir desses gráficos que a matriz do nosso modelo possui um erro numérico significativamente menor comparado a extrapolação com pontos equidistantes.
